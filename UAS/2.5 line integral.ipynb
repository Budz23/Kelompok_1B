**2.5 Integral Garis dan Ketidaktergantungan Jalur**

***2.5.1 Line Integrals***

2.5.1.1 Line (Path or Curve) Integral of a Scalar Function on Plane (Parametric Definition)





Definisi: Integral garis dari fungsi skalar $f(x, y)$ sepanjang kurva mulus $C$ di bidang yang diparameterisasi oleh $x = x(t)$ dan $y = y(t)$ untuk $a \leq t \leq b$ dihitung sebagai:  
$[ \int_C f(x, y) , dx = \int_a^b f(x(t), y(t)) x'(t) , dt ] [ \int_C f(x, y) , dy = \int_a^b f(x(t), y(t)) y'(t) , dt ] [ \int_C f(x, y) , dl = \int_a^b f(x(t), y(t)) \sqrt{(x'(t))^2 + (y'(t))^2} , dt ]$




Contoh (E2.12.2): Evaluasi $\int_C x^2 y , dx$ di mana $C$ diparameterisasi oleh $x(t) = t$ dan $y(t) = t^2$ untuk $0 \leq t \leq 1$: $[ \int_C x^2 y , dx = \int_0^1 (t^2 \cdot t^2) \cdot 1 , dt = \int_0^1 t^4 , dt = \left[ \frac{t^5}{5} \right]_0^1 = \frac{1}{5} ]$



Problem: Evaluasi $\int_C x y^2 , dy$ di mana $C$ didefinisikan oleh $x(t) = \cos t$, $y(t) = \sin t$ untuk $0 \leq t \leq \frac{\pi}{2}$. (Jawaban: $\frac{1}{4}$.)

2.5.1.2 Line Integral of a Scalar Function $f(x, y)$ along a Curve $C$ Defined by an Explicit Function





Definisi: Jika kurva $C$ didefinisikan oleh $y = g(x)$ untuk $a \leq x \leq b$, integral garis dihitung sebagai: $[ \int_C f(x, y) , dx = \int_a^b f(x, g(x)) , dx ] [ \int_C f(x, y) , dy = \int_a^b f(x, g(x)) g'(x) , dx ] [ \int_C f(x, y) , dl = \int_a^b f(x, g(x)) \sqrt{1 + (g'(x))^2} , dx ]$



Contoh (E2.12.4): Evaluasi $\int_C x^2 \cos(y) , dy$ di mana $C$ didefinisikan oleh $y = \sin(x)$ untuk $0 \leq x \leq \frac{\pi}{2}$: $[ \int_C x^2 \cos(y) , dy = \int_0^{\pi/2} x^2 \cos(\sin(x)) \cos(x) , dx ]$



Problem: Evaluasi $\int_C y^2 , dx$ di mana $C$ didefinisikan oleh $y = x^2$ untuk $0 \leq x \leq 1$. (Jawaban: $\frac{1}{3}$.)

2.5.1.3 Integral Garis Fungsi Vektor di Ruang





Definisi: Untuk $\mathbf{F} = F_x \mathbf{u}_x + F_y \mathbf{u}_y + F_z \mathbf{u}_z$ sepanjang $C$ dengan $x(t), y(t), z(t)$ untuk $a \leq t \leq b$: $[ \int_C \mathbf{F} \cdot d\mathbf{l} = \int_a^b \left( F_x \frac{dx}{dt} + F_y \frac{dy}{dt} + F_z \frac{dz}{dt} \right) dt ]$



Contoh (2.16): Evaluasi $\int_C \mathbf{F} \cdot d\mathbf{l}$ di mana $\mathbf{F} = \frac{2(x \mathbf{u}_x + y \mathbf{u}_y + z \mathbf{u}_z)}{(x^2 + y^2 + z^2)^{3/2}}$ dan $C$ dari $(1,0,0)$ ke $(0,1,0)$ dengan $x(t) = 1-t$, $y(t) = t$, $z(t) = 0$ untuk $0 \leq t \leq 1$. Hasilnya $-\sqrt{2}$.



Problem: Evaluasi $\int_C \mathbf{F} \cdot d\mathbf{l}$ di mana $\mathbf{F} = y \mathbf{u}_x + x \mathbf{u}_y$ dan $C$ adalah setengah lingkaran $x^2 + y^2 = 1$, $y \geq 0$ dari $(1,0)$ ke $(-1,0)$. (Jawaban: $2\pi$.)

***2.5.2 Ketidaktergantungan Jalur***





Definisi: Medan $\mathbf{G}$ konservatif jika $\mathbf{G} = \nabla f$ dan $\nabla \times \mathbf{G} = 0$, dengan integral $\int_A^B \mathbf{G} \cdot d\mathbf{l} = f(B) - f(A)$.



Contoh (2.18): $\mathbf{G} = \frac{-y \mathbf{u}_x + x \mathbf{u}_y}{x^2}$ tidak tergantung jalur di ${x > 0}$, dengan $f(x, y) = \frac{y}{x} + C$.



Problem: Tentukan apakah $\mathbf{G} = x^2 \mathbf{u}_x + 2xy \mathbf{u}y$ konservatif dan hitung $\int{(0,0)}^{(1,1)} \mathbf{G} \cdot d\mathbf{l}$. (Jawaban: Konservatif dengan $f = x^2 y$, integral = 1.)

**2.6 Integral Ganda**

***2.6.1 Definisi dan Perhitungan***





Definisi: $\iint_R f(x, y) , dA = \int_a^b \int_{g_1(x)}^{g_2(x)} f(x, y) , dy , dx$ (Tipe I) atau $\int_c^d \int_{h_1(y)}^{h_2(y)} f(x, y) , dx , dy$ (Tipe II).



Contoh (2.20): Evaluasi $\iint_R (x + y) , dA$ di segitiga $(0,0)$, $(1,0)$, $(0,1)$: $\int_0^1 \int_0^{1-x} (x + y) , dy , dx = 0$.



Problem: Hitung $\iint_R x^2 , dA$ di persegi $0 \leq x, y \leq 1$. (Jawaban: $\frac{1}{3}$.)

***2.6.2 Pusat Massa dan Momen Inersia***





Definisi: Massa $m = \iint_R \rho(x, y) , dA$, pusat massa $(\bar{x}, \bar{y}) = \left( \frac{M_y}{m}, \frac{M_x}{m} \right)$ dengan $M_x = \iint_R y \rho(x, y) , dA$ dan $M_y = \iint_R x \rho(x, y) , dA$.



Contoh (2.21): Untuk $\rho = 1$ di $x^2 + y^2 \leq 1$, pusat massa $(0, 0)$ dan $m = \pi$.



Problem: Tentukan pusat massa dengan $\rho = x$ di $R: 0 \leq x \leq 1$, $0 \leq y \leq x$. (Jawaban: $(\frac{2}{3}, \frac{1}{2})$.)

***2.6.3 Konversi Koordinat Kartesian ke Polar***





Definisi: $x = \rho \cos \phi$, $y = \rho \sin \phi$, $dA = \rho , d\rho , d\phi$.



Contoh (2.21): $\iint_R e^{x^2 + y^2} , dA$ di $x^2 + y^2 \leq 1$: $\int_0^{2\pi} \int_0^1 e^{\rho^2} \rho , d\rho , d\phi = \pi(e - 1)$.



Problem: Hitung $\iint_R x y , dA$ di $x^2 + y^2 \leq 4$. (Jawaban: 0.)

**2.7 Teorema Green**





Definisi: $\oint_C P , dx + Q , dy = \iint_R \left( \frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y} \right) dA$, dengan $C$ kurva tertutup sederhana.



Contoh (2.22): $\oint_C -y , dx + x , dy$ pada persegi sisi 4, hasilnya 32.



Problem: Hitung $\oint_C x , dy - y , dx$ di $x^2 + y^2 = 1$. (Jawaban: $-2\pi$.)

**2.8 Integral Permukaan**

2.8.1 Luas Permukaan





Definisi: $A(S) = \iint_R \sqrt{1 + \left( \frac{\partial f}{\partial x} \right)^2 + \left( \frac{\partial f}{\partial y} \right)^2} , dA$ untuk $z = f(x, y)$.



Contoh (2.25): Luas $z = \sqrt{1 - x^2 - y^2}$ di $x^2 + y^2 \leq 1$ adalah $2\pi$.



Problem: Hitung luas $z = x + y$ di $0 \leq x, y \leq 1$. (Jawaban: $\frac{3\sqrt{2}}{2}$.)

***2.8.2 Integral Permukaan Fungsi Skalar***





Definisi: $\iint_S G(x, y, z) , dS = \iint_R G(x, y, f(x, y)) \sqrt{1 + \left( \frac{\partial f}{\partial x} \right)^2 + \left( \frac{\partial f}{\partial y} \right)^2} , dA$.



Contoh (2.27): Untuk $G = 8xy$ di $z = 2(1 - x - y)$ di $x + y \leq 1$, hasilnya 1.



Problem: Evaluasi $\iint_S x , dS$ di $z = 1 - x^2 - y^2$, $x^2 + y^2 \leq 1$. (Jawaban: $\frac{\pi}{2}$.)

***2.8.3 Integral Permukaan Fungsi Vektor***





Definisi: Fluks $\iint_S \mathbf{G} \cdot d\mathbf{s} = \iint_S \mathbf{G} \cdot \mathbf{u}_n , dS$.



Contoh (2.26): Untuk $\mathbf{G} = \frac{x \mathbf{u}_x + y \mathbf{u}_y + z \mathbf{u}_z}{(x^2 + y^2 + z^2)^{3/2}}$ di bola satuan, fluksnya $4\pi$.



Problem: Hitung fluks $\mathbf{G} = z \mathbf{u}_z$ di $z = 1 - x^2 - y^2$, $x^2 + y^2 \leq 1$. (Jawaban: $\pi$.)
