## Vector Calculus
### 2.1  DERIVATIVES
#### Definisi Turunan

Turunan dari suatu fungsi $ y = f(x) $ dari satu variabel $ x $ didefinisikan sebagai:
$$\frac{d}{dx} f(x) = \lim_{\Delta x \to 0} \frac{f(x + \Delta x) - f(x)}{\Delta x} \tag{2.1.1}$$
#### Aproksimasi Turunan
Turunan ini sering digantikan oleh salah satu dari aproksimasi berikut:

Forward difference: $$\frac{f(x + h) - f(x)}{h} \tag{2.1.2a}$$

Backward difference: $$\frac{f(x) - f(x - h)}{h} \tag{2.1.2b}$$

Central difference: $$\frac{f(x + h) - f(x - h)}{2h} \tag{2.1.2c}$$

#### Turunan Parsial
Turunan parsial dari suatu fungsi multivariabel, katakanlah $ f(x, y) $ dari dua variabel $ x $ dan $ y $, dengan hormat kepada $ x $ dan $ y $, didefinisikan sebagai:
$$\frac{\partial}{\partial x} f(x, y) = \lim_{\Delta x \to 0} \frac{f(x + \Delta x, y) - f(x, y)}{\Delta x} \tag{2.1.3a}$$
$$\frac{\partial}{\partial y} f(x, y) = \lim_{\Delta y \to 0} \frac{f(x, y + \Delta y) - f(x, y)}{\Delta y} \tag{2.1.3b}$$
#### Aturan Rantai
Jika $ y = f(u) $ adalah fungsi yang dapat diturunkan dari variabel $ u $ di mana $ u = g(x) $ adalah fungsi dari variabel lain $ x $, maka aturan rantai menyatakan bahwa turunan $ \frac{dy}{dx} $ adalah:
$$\frac{dy}{dx} = \frac{dy}{du} \cdot \frac{du}{dx} \tag{2.1.4a}$$
Jika $ z = f(x, y) $ adalah fungsi yang dapat diturunkan dari $ x $ dan $ y $ di mana $ x = g(s, t) $ dan $ y = h(s, t) $ adalah fungsi dari dua variabel lain $ s $ dan $ t $, maka aturan rantai menyatakan bahwa turunan parsial $ z $ dengan hormat kepada $ s $ dan $ t $ adalah:
$$\frac{\partial z}{\partial s} = \frac{\partial z}{\partial x} \cdot \frac{\partial x}{\partial s} + \frac{\partial z}{\partial y} \cdot \frac{\partial y}{\partial s} \tag{2.1.4b}$$
$$\frac{\partial z}{\partial t} = \frac{\partial z}{\partial x} \cdot \frac{\partial x}{\partial t} + \frac{\partial z}{\partial y} \cdot \frac{\partial y}{\partial t} \tag{2.1.4c}$$


#### Gradien

Gradien dari fungsi satu-variabel: $\nabla f(x) $ didefinisikan sebagai $\frac{df}{dx}$.

Gradien dari fungsi dua-variabel: $\nabla f(x, y) $ didefinisikan sebagai:
$$\nabla f(x, y) = \frac{\partial f}{\partial x} \mathbf{u}_x + \frac{\partial f}{\partial y} \mathbf{u}_y \tag{2.1.5a}$$

Gradien dari fungsi tiga-variabel: $\nabla f(x, y, z) $ didefinisikan sebagai:
$$\nabla f(x, y, z) = \frac{\partial f}{\partial x} \mathbf{u}_x + \frac{\partial f}{\partial y} \mathbf{u}_y + \frac{\partial f}{\partial z} \mathbf{u}_z \tag{2.1.5b}$$
di mana $\mathbf{u}_x, \mathbf{u}_y, \mathbf{u}_z$ adalah vektor unit sepanjang sumbu $x, y, z$ masing-masing.



#### Turunan Directional
Turunan directional dari $ f(x, y, z) $ dalam arah vektor unit $\mathbf{u} = a_x \mathbf{u}_x + a_y \mathbf{u}_y + a_z \mathbf{u}_z$ dengan $ |\mathbf{u}| = 1 $ adalah dot product dari gradien dan vektor unit:
$$D_{\mathbf{u}} f(x, y, z) = \lim_{h \to 0} \frac{f(x + a_x h, y + a_y h, z + a_z h) - f(x, y, z)}{h} \tag{2.1.6a}$$
$$= \frac{\partial f}{\partial x} a_x + \frac{\partial f}{\partial y} a_y + \frac{\partial f}{\partial z} a_z = \nabla f(x, y, z) \cdot \mathbf{u} \tag{2.1.6b}$$

#### Garis Singgung
Persamaan garis singgung dari kurva $ y = f(x) $ di titik $(x_0, y_0)$ dapat ditulis sebagai:
$$y - y_0 = \left[\frac{df}{dx}\right]_{x=x_0} (x - x_0) \tag{2.1.7a}$$
Bidang Singgung
Persamaan bidang singgung dari permukaan $ f(x, y, z) = c $ di titik $(x_0, y_0, z_0)$ dapat ditulis sebagai:
$$\nabla f(x_0, y_0, z_0) \cdot (x - x_0, y - y_0, z - z_0) = 0 \tag{2.1.7b}$$

#### Contoh

##### 1.Garis Singgung:
Misalkan $ f(x, y) = x^2 + y = 4 $. Di titik $(1, 3)$, gradien adalah $\nabla f = (2x, 1)$, sehingga di $(1, 3)$, $\nabla f = (2, 1)$. Persamaan garis singgung adalah:
$$2(x - 1) + 1(y - 3) = 0 \quad \text{atau} \quad 2x + y - 5 = 0 \tag{2.1}$$



```python
import numpy as np

def gradient_tangent_line(x, y):
    df_dx = 2 * x  # Turunan parsial terhadap x
    df_dy = 1      # Turunan parsial terhadap y
    return np.array([df_dx, df_dy])

# Titik (x0, y0)
x0, y0 = 1, 3
gradient_at_point = gradient_tangent_line(x0, y0)
# Persamaan garis singgung: 2(x - x0) + 1(y - y0) = 0
# Sederhanakan: 2x + y - 5 = 0
print(f"Garis Singgung di ({x0}, {y0}): 2x + y - 5 = 0")
print(f"Gradien di titik ({x0}, {y0}): {gradient_at_point}")
```

##### 2.Bidang Singgung:
Misalkan $ f(x, y, z) = x^2 - y^2 + 4z = 0 $. Di titik $(1, 1, 0)$, gradien adalah $\nabla f = (2x, -2y, 4)$, sehingga di $(1, 1, 0)$, $\nabla f = (2, -2, 4)$. Persamaan bidang singgung adalah:
$$2(x - 1) - 2(y - 1) + 4(z - 0) = 0 \quad \text{atau} \quad 2x - 2y + 4z - 4 = 0 \tag{2.2}$$

```python
import numpy as np

def gradient_tangent_plane(x, y, z):
    df_dx = 2 * x  # Turunan parsial terhadap x
    df_dy = -2 * y # Turunan parsial terhadap y
    df_dz = 4      # Turunan parsial terhadap z
    return np.array([df_dx, df_dy, df_dz])

# Titik (x0, y0, z0)
x0, y0, z0 = 1, 1, 0
gradient_at_point = gradient_tangent_plane(x0, y0, z0)
# Persamaan bidang singgung: 2(x - x0) - 2(y - y0) + 4(z - z0) = 0
# Sederhanakan: 2x - 2y + 4z - 4 = 0 atau x - y + 2z = 0
print(f"Bidang Singgung di ({x0}, {y0}, {z0}): x - y + 2z = 0")
print(f"Gradien di titik ({x0}, {y0}, {z0}): {gradient_at_point}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Plot tangent line to x^2 - y = 4
x = np.linspace(-5, 5, 400)
y = x**2 - 4
fig, ax = plt.subplots()
ax.plot(x, y, label=r'$x^2 - y = 4$')

# Titik P = (-2, 0)
x0, y0 = -2, 0
ax.plot(x0, y0, 'ro', label='P = (-2, 0)')

# Gradien di P = (2x, -1)
grad_x = 2 * x0
grad_y = -1
ax.quiver(x0, y0, grad_x, grad_y, color='r', scale=20, label='Gradient at P')

# Garis singgung (aproksimasi sederhana)
m = grad_x / -grad_y if grad_y != 0 else 0
y_tangent = y0 + m * (x - x0)
ax.plot(x, y_tangent, 'r--', label='Tangent line')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
ax.grid(True)
ax.set_title('Gradient and Tangent Line to $x^2 - y = 4$')
plt.show()
```

#### Contoh
Gradien Negatif dari Potensial Listrik untuk Mendapatkan Medan Listrik
Medan listrik dipotensial $ V $ yang dihasilkan oleh dua muatan listrik berdasar yang sama, $ +Q $ dan $ -Q $, masing-masing terletak di $(0, 0, 0)$ dan $(0, 4, 0)$, masing-masing memiliki magnitudo dan arah yang berlawanan, dapat dinyatakan sebagai:
$$\vec{E} = -\frac{Q}{4 \pi \epsilon_0} \left[ \frac{\vec{r}_1}{R_1^3} - \frac{\vec{r}_2}{R_2^3} \right], \quad \vec{r}_1 = (x, y, z), \quad \vec{r}_2 = (x, y-4, z)$$
Medan listrik $ \vec{E} $ dapat diperoleh sebagai gradien negatif dari $ V $:
$$\vec{E} = -\nabla V(x, y, z) = -u_x \frac{\partial V}{\partial x} - u_y \frac{\partial V}{\partial y}$$
$$\frac{\partial^2 V}{\partial x^2} = \frac{Q}{4 \pi \epsilon_0} \left[ \frac{(y-4)^2}{(x^2 + (y-4)^2)^{3/2}} - \frac{y^2}{(x^2 + y^2)^{3/2}} \right] u_x$$

Program MATLAB "em02e02.m" (telah diubah ke pyhton) telah dijalankan untuk membuat plot $ V(x, y) $ dan $ \vec{E} $ seperti yang ditunjukkan pada Gambar 2.2(a) dan (b) masing-masing.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter
Q = 1.0  # Muatan (dalam satuan arbitrer)
epsilon_0 = 1.0  # Permitivitas (dalam satuan arbitrer)
k = 1 / (4 * np.pi * epsilon_0)  # Konstanta Coulomb

# Grid
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

# Posisi muatan
r1 = np.sqrt(X**2 + Y**2)  # Jarak ke muatan +Q di (0, 0)
r2 = np.sqrt(X**2 + (Y - 4)**2)  # Jarak ke muatan -Q di (0, 4)

# Potensial listrik V(x, y)
V = k * Q * (1 / r1 - 1 / r2)

# Plot kontur potensial
plt.figure()
contour = plt.contour(X, Y, V, levels=20, cmap='viridis')
plt.clabel(contour, inline=True, fontsize=8)
plt.title('Potensial Listrik $V(x, y)$')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.colorbar(label='Potensial (V)')
plt.show()
```

## 2.2 VECTOR FUNCTIONS

### 1. Definisi Fungsi Vektor

Fungsi vektor dalam 2D dan 3D dituliskan sebagai:

- **2D:**

$$
\mathbf{p}(t) = x(t)\mathbf{u}_x + y(t)\mathbf{u}_y \tag{2.2.1a}
$$

**3D:**

$$
\mathbf{p}(t) = x(t)\mathbf{u}_x + y(t)\mathbf{u}_y + z(t)\mathbf{u}_z \tag{2.2.1b}
$$

**Keterangan:**
- $\mathbf{p}(t)$: Vektor posisi sebagai fungsi waktu
- $x(t), y(t), z(t)$: Komponen skalar dalam arah masing-masing sumbu
- $\mathbf{u}_x, \mathbf{u}_y, \mathbf{u}_z$: Vektor satuan arah x, y, z

---

### 2. Vektor Tangen

Vektor tangen adalah turunan dari posisi vektor $\mathbf{p}(t)$ terhadap waktu $t$:

- **2D:**

$$
\mathbf{p}'(t) = \frac{d\mathbf{p}(t)}{dt} = \frac{dx(t)}{dt} \mathbf{u}_x + \frac{dy(t)}{dt} \mathbf{u}_y \tag{2.2.2a}
$$

- **3D:**

$$
\mathbf{p}'(t) = \frac{dx(t)}{dt} \mathbf{u}_x + \frac{dy(t)}{dt} \mathbf{u}_y + \frac{dz(t)}{dt} \mathbf{u}_z \tag{2.2.2b}
$$

---

### 3. Panjang Lintasan (Arc Length)

Panjang lintasan dari $t_0$ ke $t_1$ diberikan oleh:

$$
l = \int_{t_0}^{t_1} \left| \mathbf{p}'(t) \right| dt = \int_{t_0}^{t_1} \sqrt{(x'(t))^2 + (y'(t))^2 + (z'(t))^2} \, dt \tag{2.2.3}
$$

---

### 4. Parameterisasi Berdasarkan Panjang Lintasan

$$
s = \int_{t_0}^{t} \left| \mathbf{p}'(\tau) \right| d\tau = h(t) \tag{2.2.4}
$$

---

### 5. Aturan Turunan Fungsi Vektor

Beberapa aturan turunan untuk fungsi vektor $\mathbf{p}(t), \mathbf{q}(t)$ dan skalar $\phi(t)$:

- **a. Perkalian skalar dan vektor:**

$$
\frac{d}{dt}[\phi(t)\mathbf{p}(t)] = \phi'(t)\mathbf{p}(t) + \phi(t)\mathbf{p}'(t) \tag{2.2.5a}
$$

- **b. Penjumlahan vektor:**

$$
\frac{d}{dt}[\mathbf{p}(t) + \mathbf{q}(t)] = \mathbf{p}'(t) + \mathbf{q}'(t) \tag{2.2.5b}
$$

- **c. Dot product:**

$$
\frac{d}{dt}[\mathbf{p}(t) \cdot \mathbf{q}(t)] = \mathbf{p}'(t) \cdot \mathbf{q}(t) + \mathbf{p}(t) \cdot \mathbf{q}'(t) \tag{2.2.5c}
$$

- **d. Cross product:**

$$
\frac{d}{dt}[\mathbf{p}(t) \times \mathbf{q}(t)] = \mathbf{p}'(t) \times \mathbf{q}(t) + \mathbf{p}(t) \times \mathbf{q}'(t) \tag{2.2.5d}
$$

---

### 6. Contoh: Circular Helix

Fungsi vektor heliks:

$$
\mathbf{p}(t) = t\mathbf{u}_x + \cos(2\pi t)\mathbf{u}_y + \sin(2\pi t)\mathbf{u}_z \tag{E2.3.1}
$$

#### a. Vektor Tangen:

$$
\mathbf{p}'(t) = \mathbf{u}_x - 2\pi \sin(2\pi t)\mathbf{u}_y + 2\pi \cos(2\pi t)\mathbf{u}_z \tag{E2.3.2}
$$

#### b. Panjang Lintasan dari $t = 0$ ke $t = 1$:

$$
l = \int_0^1 \sqrt{1 + 4\pi^2} \, dt = \sqrt{1 + 4\pi^2} \approx 6.3623 \tag{E2.3.3}
$$

#### c. Parameterisasi Berdasarkan Panjang Lintasan:

Jika $t = \frac{l}{\sqrt{1 + 4\pi^2}}$, maka:

$$
\mathbf{p}(l) = \frac{l}{\sqrt{1 + 4\pi^2}} \mathbf{u}_x + \cos\left(\frac{2\pi l}{\sqrt{1 + 4\pi^2}}\right) \mathbf{u}_y + \sin\left(\frac{2\pi l}{\sqrt{1 + 4\pi^2}}\right) \mathbf{u}_z \tag{E2.3.6}
$$

#### d. Vektor Tangen Satuan:

$$
\mathbf{p}'(l) = \frac{1}{\sqrt{1 + 4\pi^2}} \mathbf{u}_x - \frac{2\pi}{\sqrt{1 + 4\pi^2}} \sin\left(\frac{2\pi l}{\sqrt{1 + 4\pi^2}}\right) \mathbf{u}_y + \frac{2\pi}{\sqrt{1 + 4\pi^2}} \cos\left(\frac{2\pi l}{\sqrt{1 + 4\pi^2}}\right) \mathbf{u}_z \tag{E2.3.7}
$$



## Contoh 2.3 – Tangent Line dan Panjang Busur dari Circular Helix

Diberikan sebuah fungsi vektor dalam 3D sebagai berikut:

$$
\mathbf{p}(t) = t\mathbf{u}_x + \cos(2\pi t)\mathbf{u}_y + \sin(2\pi t)\mathbf{u}_z \tag{E2.3.1}
$$

Turunan dari vektor tersebut (vektor tangen):

$$
\mathbf{p}'(t) = \frac{d\mathbf{p}}{dt} = \mathbf{u}_x - 2\pi \sin(2\pi t)\mathbf{u}_y + 2\pi \cos(2\pi t)\mathbf{u}_z \tag{E2.3.2}
$$

Untuk $t = 1$, kita akan menghitung vektor posisi $\mathbf{p}(1)$ dan vektor tangen $\mathbf{p}'(1)$ kemudian menampilkannya pada plot.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Parameter t
t = np.linspace(0, 1, 500)

# Posisi pada kurva heliks
x = t
y = np.cos(2 * np.pi * t)
z = np.sin(2 * np.pi * t)

# Titik t = 1 untuk vektor tangen
t0 = 1.0
p0 = np.array([t0, np.cos(2 * np.pi * t0), np.sin(2 * np.pi * t0)])

# Vektor tangen p'(t)
tangent_vector = np.array([1, -2 * np.pi * np.sin(2 * np.pi * t0), 2 * np.pi * np.cos(2 * np.pi * t0)])

# Panjang segmen vektor tangen (misal 0.2 seperti di MATLAB)
segment_length = 0.2
tangent_scaled = tangent_vector * segment_length

# Plot
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')

# Kurva heliks
ax.plot(x, y, z, label='Helix Curve', color='blue')

# Titik p(t=1)
ax.scatter(*p0, color='red', s=50, label='p(t=1)')

# Vektor tangen dari p(t=1)
ax.quiver(
    p0[0], p0[1], p0[2], 
    tangent_scaled[0], tangent_scaled[1], tangent_scaled[2],
    color='red', length=1.0, normalize=False, label='Tangent Vector (Δt = 0.2)'
)

# Label dan tampilan
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title('Circular Helix and Tangent Vector at $t=1$')
ax.legend()
ax.grid(True)

plt.show()
```

### 2.3 VELOCITY AND ACCELERATION
Kecepatan dan percepatan adalah konsep penting dalam kalkulus vektor yang digunakan untuk menggambarkan gerak suatu partikel berdasarkan posisinya terhadap waktu. Jika posisi partikel diberikan oleh vektor posisi $\mathbf{p}(t) = x(t)\mathbf{u}_x + y(t)\mathbf{u}_y + z(t)\mathbf{u}_z$,

1. Rumus Kecepatan dan Percepatan Dasar

Kecepatan ($\mathbf{v}(t)$):
$$\mathbf{v}(t) = \frac{d}{dt} \mathbf{p}(t) = \frac{dx}{dt} \mathbf{u}_x + \frac{dy}{dt} \mathbf{u}_y + \frac{dz}{dt} \mathbf{u}_z$$

$\mathbf{p}(t)$: Vektor posisi sebagai fungsi waktu $t$.

$\frac{d}{dt}$: Turunan terhadap waktu $t$.

$x(t), y(t), z(t)$: Komponen posisi di sumbu $x$, $y$, dan $z$.

$\mathbf{u}_x, \mathbf{u}_y, \mathbf{u}_z$: Vektor satuan untuk sumbu $x$, $y$, dan $z$.


Percepatan ($\mathbf{a}(t)$):
$$\mathbf{a}(t) = \frac{d^2}{dt^2} \mathbf{p}(t) = \frac{d^2x}{dt^2} \mathbf{u}_x + \frac{d^2y}{dt^2} \mathbf{u}_y + \frac{d^2z}{dt^2} \mathbf{u}_z$$

$\frac{d^2}{dt^2}$: Turunan kedua terhadap waktu $t$.


Magnitudo Kecepatan:
$$\|\mathbf{v}(t)\|^2 = \left(\frac{dx}{dt}\right)^2 + \left(\frac{dy}{dt}\right)^2 + \left(\frac{dz}{dt}\right)^2$$

$\|\mathbf{v}(t)\|$: Besar (magnitudo) vektor kecepatan, juga disebut laju.


Laju sebagai Turunan Panjang Kurva:
$$v(t) = \frac{d}{dt} \int \|\mathbf{p}'(t)\| dt$$

$\|\mathbf{p}'(t)\|$: Magnitudo vektor kecepatan instan.


Kondisi Laju Konstan:
$$\|\mathbf{v}\|^2 = v \cdot v = c^2$$

$v$: Kecepatan skalar (laju).
$c$: Konstanta kecepatan (jika laju konstan).



2. Gerak dengan Laju Konstan (Lingkaran)

Posisi:
$$\mathbf{p}(t) = R \cos(\omega t) \mathbf{u}_x + R \sin(\omega t) \mathbf{u}_y$$

$R$: Radius lingkaran.

$\omega$: Kecepatan sudut (rad/s).

$\cos(\omega t), \sin(\omega t)$: Fungsi trigonometri yang menentukan posisi pada lingkaran.


Kecepatan:
$$\mathbf{v}(t) = -R \omega \sin(\omega t) \mathbf{u}_x + R \omega \cos(\omega t) \mathbf{u}_y$$

Percepatan:
$$\mathbf{a}(t) = -R \omega^2 \cos(\omega t) \mathbf{u}_x - R \omega^2 \sin(\omega t) \mathbf{u}_y = -\omega^2 \mathbf{p}(t)$$

$\omega^2$: Besaran percepatan sentripetal.



3. Vektor Satuan dan Kelengkungan

Vektor Satuan Tanggen ($\mathbf{u}_T$):
$$\mathbf{u}_T = \frac{\mathbf{v}(t)}{\|\mathbf{v}(t)\|}$$
atau
$$\mathbf{u}_T = \frac{\mathbf{p}'(t)}{\|\mathbf{p}'(t)\|}$$

$\mathbf{u}_T$: Vektor satuan tangen, arah gerak partikel.


Kelengkungan ($\kappa$):
$$\kappa = \frac{\|\mathbf{u}_T'\|}{\|\mathbf{v}(t)\|}$$

$\|\mathbf{u}_T'\|$: Laju perubahan vektor satuan tangen.

$\kappa$: Ukuran kelengkungan kurva.


Radius Kelengkungan ($r_k$):
$$r_k = \frac{1}{\kappa}$$

$r_k$: Jari-jari lingkaran kelengkungan.





4. Contoh Lintasan Peluru

Posisi:
$$\mathbf{p}(t) = (v_0 \cos \theta \cdot t) \mathbf{u}_x + (v_0 \sin \theta \cdot t - \frac{1}{2} g t^2) \mathbf{u}_y$$

$v_0$: Kecepatan awal.

$\theta$: Sudut peluncuran.

$g$: Percepatan gravitasi ($\approx 9,8 \, \text{m/s}^2$).




Waktu Dampak:
$$t = \frac{2 v_0 \sin \theta}{g}$$

Kecepatan dan Posisi Dampak:
$$\mathbf{p}(t_f) = (v_0 \cos \theta \cdot t_f) \mathbf{u}_x$$
$$\mathbf{v}(t_f) = [v_0 \cos \theta \mathbf{u}_x + (v_0 \sin \theta - g t_f) \mathbf{u}_y]$$

$t_f$: Waktu dampak.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter
t = np.linspace(0, 2*np.pi, 100)
omega = 1  # Kecepatan sudut
x = np.cos(omega * t)
y = np.sin(omega * t)

# Titik untuk vektor tangen dan normal (t = 0 dan t = pi/2)
t_points = [0, np.pi/2]
x_points = np.cos(omega * t_points)
y_points = np.sin(omega * t_points)

# Hitung vektor tangen (turunan)
dt = t[1] - t[0]
tangent_x = -np.sin(omega * t_points)
tangent_y = np.cos(omega * t_points)
tangent_magnitude = np.sqrt(tangent_x**2 + tangent_y**2)
uT_x = tangent_x / tangent_magnitude
uT_y = tangent_y / tangent_magnitude

# Hitung vektor normal (tegak lurus tangen, rotasi 90 derajat)
uN_x = -uT_y
uN_y = uT_x

# Lingkaran kelengkungan (radius kelengkungan = 1 untuk elips ini pada titik tertentu)
radius_curvature = 1
theta = np.linspace(0, 2*np.pi, 100)
for i in range(len(t_points)):
    x_circle = x_points[i] + radius_curvature * np.cos(theta) * uN_x[i]
    y_circle = y_points[i] + radius_curvature * np.sin(theta) * uN_y[i]
    plt.plot(x_circle, y_circle, 'r--', alpha=0.5)

# Plot elips
plt.plot(x, y, label='Elips')
plt.plot(x_points, y_points, 'ko')  # Titik observasi

# Plot vektor tangen dan normal
scale = 0.5  # Skala vektor untuk visualisasi
for i in range(len(t_points)):
    plt.quiver(x_points[i], y_points[i], uT_x[i] * scale, uT_y[i] * scale, color='g', label='Tangent vector' if i == 0 else "", angles='xy', scale_units='xy', scale=1)
    plt.quiver(x_points[i], y_points[i], uN_x[i] * scale, uN_y[i] * scale, color='c', label='Normal vector' if i == 0 else "", angles='xy', scale_units='xy', scale=1)

# Label dan gaya
plt.title('Unit Tangent Vectors and Circles of Curvature at points (2, 0) and (0, 1) for Example 2.5')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axis('equal')
plt.legend()
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.xlim(-1.5, 1.5)
plt.ylim(-1.5, 1.5)
plt.show()
```


## 2.4 Divergensi dan Rotasi (Curl)

### 2.4.1 Definisi

**Divergensi** dari medan vektor **F** (dijelaskan oleh fungsi vektor dua atau tiga variabel) pada suatu titik observasi didefinisikan sebagai:

$$\text{div } \mathbf{F} = \lim_{\Delta V \to 0} \frac{\oint_{\partial V} \mathbf{F} \cdot d\mathbf{S}}{\Delta V} \quad \text{(2.4.1)}$$

**Keterangan:**
* $\Delta V$: Volume diferensial yang mengandung titik observasi.
* $\partial V$: Permukaan batas tertutup yang mengelilingi volume $\Delta V$.
* $d\mathbf{S}$: Vektor permukaan diferensial keluar, dengan $dS$ adalah luas permukaan dan $\mathbf{n}$ adalah vektor normal satuan keluar.
* $\oint_{\partial V} \mathbf{F} \cdot d\mathbf{S}$: Integral permukaan dari $\mathbf{F}$ di atas permukaan batas tertutup $\partial V$.
* **Makna Fisik:** Ini adalah batas fluks bersih keluar dari $\mathbf{F}$ per unit volume saat volume mendekati nol. Divergensi mengukur "sumber" atau "penyerap" suatu medan vektor pada suatu titik.

**Rotasi (curl)** dari medan vektor **F** pada suatu titik observasi didefinisikan sebagai:

$$\text{curl } \mathbf{F} = \lim_{\Delta S \to 0} \frac{1}{\Delta S} \oint_C \mathbf{F} \cdot d\mathbf{r} \quad \text{(2.4.2)}$$

**Keterangan:**
* $\Delta S$: Luas permukaan bidang diferensial yang mengandung titik observasi.
* $C$: Garis batas tertutup dari permukaan bidang diferensial.
* $d\mathbf{r}$: Vektor elemen garis diferensial.
* $\oint_C \mathbf{F} \cdot d\mathbf{r}$: Integral garis dari $\mathbf{F}$ di sepanjang garis batas tertutup $C$ (sirkulasi).
* **Makna Fisik:** Ini adalah batas sirkulasi bersih (rotasi) dari $\mathbf{F}$ per unit area saat area mendekati nol. Curl mengukur kecenderungan medan vektor untuk berotasi atau berputar di sekitar suatu titik. Arah vektor curl menunjukkan sumbu rotasi, dan magnitudenya menunjukkan sirkulasi maksimum per unit area.

### 2.4.2 Sistem Koordinat Kartesian dengan $(x, y, z)$

Untuk medan vektor $\mathbf{F} = F_x \mathbf{u}_x + F_y \mathbf{u}_y + F_z \mathbf{u}_z$:

**Gradien** (untuk fungsi skalar $\phi(x,y,z)$):
$$\nabla \phi = \frac{\partial \phi}{\partial x} \mathbf{u}_x + \frac{\partial \phi}{\partial y} \mathbf{u}_y + \frac{\partial \phi}{\partial z} \mathbf{u}_z \quad \text{(2.4.3a)}$$

**Divergensi:**
$$\nabla \cdot \mathbf{F} = \frac{\partial F_x}{\partial x} + \frac{\partial F_y}{\partial y} + \frac{\partial F_z}{\partial z} \quad \text{(2.4.4a)}$$

**Rotasi:**
$$\nabla \times \mathbf{F} = \left( \frac{\partial F_z}{\partial y} - \frac{\partial F_y}{\partial z} \right) \mathbf{u}_x + \left( \frac{\partial F_x}{\partial z} - \frac{\partial F_z}{\partial x} \right) \mathbf{u}_y + \left( \frac{\partial F_y}{\partial x} - \frac{\partial F_x}{\partial y} \right) \mathbf{u}_z \quad \text{(2.4.5a)}$$
Bisa juga dalam bentuk determinan:
$$\nabla \times \mathbf{F} = \begin{vmatrix} \mathbf{u}_x & \mathbf{u}_y & \mathbf{u}_z \\ \frac{\partial}{\partial x} & \frac{\partial}{\partial y} & \frac{\partial}{\partial z} \\ F_x & F_y & F_z \end{vmatrix}$$

**Laplacian** (untuk fungsi skalar $\phi(x,y,z)$):
$$\nabla^2 \phi = \frac{\partial^2 \phi}{\partial x^2} + \frac{\partial^2 \phi}{\partial y^2} + \frac{\partial^2 \phi}{\partial z^2} \quad \text{(2.4.6a)}$$

### 2.4.3 Sistem Koordinat Silinder dengan $(r, \theta, z)$

Untuk medan vektor $\mathbf{F} = F_r \mathbf{u}_r + F_\theta \mathbf{u}_\theta + F_z \mathbf{u}_z$:

**Gradien** (untuk fungsi skalar $\phi(r, \theta, z)$):
$$\nabla \phi = \frac{\partial \phi}{\partial r} \mathbf{u}_r + \frac{1}{r} \frac{\partial \phi}{\partial \theta} \mathbf{u}_\theta + \frac{\partial \phi}{\partial z} \mathbf{u}_z \quad \text{(2.4.3b)}$$

**Divergensi:**
$$\nabla \cdot \mathbf{F} = \frac{1}{r} \frac{\partial}{\partial r} (r F_r) + \frac{1}{r} \frac{\partial F_\theta}{\partial \theta} + \frac{\partial F_z}{\partial z} \quad \text{(2.4.4b)}$$

**Rotasi:**
$$\nabla \times \mathbf{F} = \left( \frac{1}{r} \frac{\partial F_z}{\partial \theta} - \frac{\partial F_\theta}{\partial z} \right) \mathbf{u}_r + \left( \frac{\partial F_r}{\partial z} - \frac{\partial F_z}{\partial r} \right) \mathbf{u}_\theta + \frac{1}{r} \left( \frac{\partial}{\partial r} (r F_\theta) - \frac{\partial F_r}{\partial \theta} \right) \mathbf{u}_z \quad \text{(2.4.5b)}$$

**Laplacian** (untuk fungsi skalar $\phi(r, \theta, z)$):
$$\nabla^2 \phi = \frac{1}{r} \frac{\partial}{\partial r} \left( r \frac{\partial \phi}{\partial r} \right) + \frac{1}{r^2} \frac{\partial^2 \phi}{\partial \theta^2} + \frac{\partial^2 \phi}{\partial z^2} \quad \text{(2.4.6b)}$$

### 2.4.4 Sistem Koordinat Bola dengan $(R, \theta, \phi)$

Untuk medan vektor $\mathbf{F} = F_R \mathbf{u}_R + F_\theta \mathbf{u}_\theta + F_\phi \mathbf{u}_\phi$:

**Gradien** (untuk fungsi skalar $\phi(R, \theta, \phi)$):
$$\nabla \phi = \frac{\partial \phi}{\partial R} \mathbf{u}_R + \frac{1}{R} \frac{\partial \phi}{\partial \theta} \mathbf{u}_\theta + \frac{1}{R \sin \theta} \frac{\partial \phi}{\partial \phi} \mathbf{u}_\phi \quad \text{(2.4.3c)}$$

**Divergensi:**
$$\nabla \cdot \mathbf{F} = \frac{1}{R^2} \frac{\partial}{\partial R} (R^2 F_R) + \frac{1}{R \sin \theta} \frac{\partial}{\partial \theta} (\sin \theta F_\theta) + \frac{1}{R \sin \theta} \frac{\partial F_\phi}{\partial \phi} \quad \text{(2.4.4c)}$$

**Rotasi:**
$$\nabla \times \mathbf{F} = \frac{1}{R \sin \theta} \left( \frac{\partial}{\partial \theta} (\sin \theta F_\phi) - \frac{\partial F_\theta}{\partial \phi} \right) \mathbf{u}_R + \frac{1}{R} \left( \frac{1}{\sin \theta} \frac{\partial F_R}{\partial \phi} - \frac{\partial}{\partial R} (R F_\phi) \right) \mathbf{u}_\theta + \frac{1}{R} \left( \frac{\partial}{\partial R} (R F_\theta) - \frac{\partial F_R}{\partial \theta} \right) \mathbf{u}_\phi \quad \text{(2.4.5c)}$$

**Laplacian** (untuk fungsi skalar $\phi(R, \theta, \phi)$):
$$\nabla^2 \phi = \frac{1}{R^2} \frac{\partial}{\partial R} \left( R^2 \frac{\partial \phi}{\partial R} \right) + \frac{1}{R^2 \sin \theta} \frac{\partial}{\partial \theta} \left( \sin \theta \frac{\partial \phi}{\partial \theta} \right) + \frac{1}{R^2 \sin^2 \theta} \frac{\partial^2 \phi}{\partial \phi^2} \quad \text{(2.4.6c)}$$



```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# --- 1. Definisikan Medan Vektor F(x, y, z) = xy^2 i + yz^2 j + zx^2 k ---
def F(x, y, z):
    """
    Mendefinisikan medan vektor F = (xy^2, yz^2, zx^2).
    Mengembalikan komponen Fx, Fy, Fz.
    """
    Fx = x * y**2
    Fy = y * z**2
    Fz = z * x**2
    return Fx, Fy, Fz

# --- 2. Fungsi untuk Menghitung Divergensi ---
def calculate_divergence(Fx, Fy, Fz, dx, dy, dz):
    """
    Menghitung divergensi (nabla . F) secara numerik.
    Menggunakan perbedaan hingga (finite difference) untuk turunan parsial.
    """
    # Turunan parsial
    dFx_dx = np.gradient(Fx, dx, axis=0) # Turunan Fx terhadap x
    dFy_dy = np.gradient(Fy, dy, axis=1) # Turunan Fy terhadap y
    dFz_dz = np.gradient(Fz, dz, axis=2) # Turunan Fz terhadap z

    # Divergensi = dFx/dx + dFy/dy + dFz/dz
    return dFx_dx + dFy_dy + dFz_dz

# --- 3. Fungsi untuk Menghitung Rotasi (Curl) ---
def calculate_curl(Fx, Fy, Fz, dx, dy, dz):
    """
    Menghitung rotasi (nabla x F) secara numerik.
    Menggunakan perbedaan hingga (finite difference) untuk turunan parsial.
    Mengembalikan komponen curl_x, curl_y, curl_z.
    """
    # Turunan parsial yang dibutuhkan untuk curl
    dFz_dy = np.gradient(Fz, dy, axis=1)
    dFy_dz = np.gradient(Fy, dz, axis=2)

    dFx_dz = np.gradient(Fx, dz, axis=2)
    dFz_dx = np.gradient(Fz, dx, axis=0)

    dFy_dx = np.gradient(Fy, dx, axis=0)
    dFx_dy = np.gradient(Fx, dy, axis=1)

    # Komponen-komponen curl
    curl_x = dFz_dy - dFy_dz
    curl_y = dFx_dz - dFz_dx
    curl_z = dFy_dx - dFx_dy
    return curl_x, curl_y, curl_z

# --- 4. Siapkan Grid untuk Plotting ---
# Rentang untuk x, y, z
x_range = np.linspace(-2, 2, 20) # Mengurangi jumlah poin untuk visualisasi yang lebih jelas
y_range = np.linspace(-2, 2, 20)
z_range = np.linspace(-2, 2, 20)

# Ukuran langkah untuk perhitungan turunan
dx = x_range[1] - x_range[0]
dy = y_range[1] - y_range[0]
dz = z_range[1] - z_range[0]

# Buat meshgrid 3D
X, Y, Z = np.meshgrid(x_range, y_range, z_range, indexing='ij')

# Hitung komponen medan vektor F pada setiap titik di grid
Fx_grid, Fy_grid, Fz_grid = F(X, Y, Z)

# Hitung divergensi dan rotasi secara numerik
div_grid = calculate_divergence(Fx_grid, Fy_grid, Fz_grid, dx, dy, dz)
curl_x_grid, curl_y_grid, curl_z_grid = calculate_curl(Fx_grid, Fy_grid, Fz_grid, dx, dy, dz)

# --- 5. Visualisasi ---

# Buat figure dan subplots
fig = plt.figure(figsize=(18, 6))

# --- Plot 1: Medan Vektor (pada bidang z=0) ---
ax1 = fig.add_subplot(131)
# Pilih slice z=0 untuk visualisasi 2D
z_slice_idx = np.argmin(np.abs(z_range - 0)) # Cari indeks z terdekat ke 0

# Ambil data untuk slice 2D
X_2d = X[:, :, z_slice_idx]
Y_2d = Y[:, :, z_slice_idx]
Fx_2d = Fx_grid[:, :, z_slice_idx]
Fy_2d = Fy_grid[:, :, z_slice_idx]

# Normalisasi panah untuk visualisasi yang lebih baik
magnitude = np.sqrt(Fx_2d**2 + Fy_2d**2)
# Hindari pembagian dengan nol
Fx_2d_norm = np.where(magnitude != 0, Fx_2d / magnitude, 0)
Fy_2d_norm = np.where(magnitude != 0, Fy_2d / magnitude, 0)

# Plot quiver (vektor)
ax1.quiver(X_2d, Y_2d, Fx_2d_norm, Fy_2d_norm, magnitude, cmap='viridis', scale=40, width=0.003)
ax1.set_title('Medan Vektor F (pada z=0)')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_aspect('equal')
ax1.grid(True)


# --- Plot 2: Peta Panas Divergensi (pada bidang z=0) ---
ax2 = fig.add_subplot(132)
div_2d = div_grid[:, :, z_slice_idx] # Ambil divergensi pada slice z=0
c2 = ax2.pcolormesh(X_2d, Y_2d, div_2d, cmap='coolwarm', shading='auto')
fig.colorbar(c2, ax=ax2, label='Divergensi F')
ax2.set_title('Peta Panas Divergensi (div F) pada z=0')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_aspect('equal')


# --- Plot 3: Peta Panas Komponen Z dari Rotasi (Curl_z) (pada bidang z=0) ---
ax3 = fig.add_subplot(133)
curl_z_2d = curl_z_grid[:, :, z_slice_idx] # Ambil komponen Z dari curl pada slice z=0
c3 = ax3.pcolormesh(X_2d, Y_2d, curl_z_2d, cmap='seismic', shading='auto')
fig.colorbar(c3, ax=ax3, label='Komponen Z dari Rotasi F')
ax3.set_title('Peta Panas Komponen Z dari Rotasi (curl_z F) pada z=0')
ax3.set_xlabel('X')
ax3.set_ylabel('Y')
ax3.set_aspect('equal')


plt.tight_layout() # Sesuaikan layout agar tidak tumpang tindih
plt.show()

# Komentar: Verifikasi nilai divergensi dan rotasi pada titik P(1, 2, 3)
# Ini adalah perhitungan simbolik dari SymPy sebelumnya, yang sekarang kita evaluasi secara numerik
x_val, y_val, z_val = 1, 2, 3

# Divergensi eksak: x^2 + y^2 + z^2
div_exact_at_P = x_val**2 + y_val**2 + z_val**2
print(f"\nDivergensi eksak F pada P({x_val}, {y_val}, {z_val}): {div_exact_at_P}")

# Komponen rotasi eksak: (-2yz)i + (-2zx)j + (-2xy)k
curl_x_exact_at_P = -2 * y_val * z_val
curl_y_exact_at_P = -2 * z_val * x_val
curl_z_exact_at_P = -2 * x_val * y_val
print(f"Komponen X Rotasi eksak F pada P({x_val}, {y_val}, {z_val}): {curl_x_exact_at_P}")
print(f"Komponen Y Rotasi eksak F pada P({x_val}, {y_val}, {z_val}): {curl_y_exact_at_P}")
print(f"Komponen Z Rotasi eksak F pada P({x_val}, {y_val}, {z_val}): {curl_z_exact_at_P}")

# Catatan: Perhitungan numerik divergensi dan rotasi menggunakan np.gradient
# mungkin akan sedikit berbeda dari hasil eksak simbolik karena
# ini adalah aproksimasi perbedaan hingga. Akurasi meningkat dengan grid yang lebih halus.
```


**Teorema Stokes:**
$$\iint_S (\nabla \times \mathbf{A}) \cdot d\mathbf{S} = \oint_C \mathbf{A} \cdot d\mathbf{l}$$
*Keterangan:* Menghubungkan integral permukaan dari curl vektor dengan integral garis dari vektor tersebut di sepanjang batas kurva tertutup.

**Keterangan:**
* $\iint_S$: Integral permukaan di atas permukaan $S$.
* $S$: Permukaan.
* $d\mathbf{S}$: Elemen vektor area diferensial permukaan, arahnya normal terhadap permukaan.
* $\oint_C$: Integral garis di sepanjang kurva tertutup $C$.
* $C$: Kurva tertutup yang membatasi permukaan $S$.
* $d\mathbf{l}$: Elemen vektor panjang diferensial di sepanjang kurva.

**Teorema Divergensi:**
$$\iiint_V (\nabla \cdot \mathbf{A}) dV = \oiint_S \mathbf{A} \cdot d\mathbf{S}$$
*Keterangan:* Menghubungkan integral volume dari divergensi vektor dengan integral permukaan dari vektor tersebut di sepanjang batas permukaan tertutup.

**Keterangan:**
* $\iiint_V$: Integral volume di atas volume $V$.
* $V$: Volume.
* $dV$: Elemen volume diferensial.
* $\oiint_S$: Integral permukaan tertutup di atas permukaan $S$ yang membatasi volume $V$.
* $S$: Permukaan tertutup yang membatasi volume $V$.
* $d\mathbf{S}$: Elemen vektor area diferensial permukaan, arahnya normal keluar dari volume.

**Contoh 2.8: Divergensi dan Curl dari Medan Listrik**

Untuk medan listrik $\mathbf{E}$ yang dihasilkan oleh muatan $q$ yang terdistribusi seragam di seluruh volume bola berjari-jari $a$:

**Untuk $r \le a$ (di dalam bola):**
$$\mathbf{E} = \frac{q}{4\pi\epsilon_0 r^3} r \mathbf{u}_r \quad \text{atau} \quad \mathbf{E} = \frac{q}{4\pi\epsilon_0 a^3} \mathbf{r}$$
*Keterangan:* Medan listrik di dalam bola, sebanding dengan vektor posisi $\mathbf{r}$.

**Keterangan:**
* $\mathbf{E}$: Medan listrik (vektor).
* $q$: Besarnya muatan listrik (skalar).
* $\epsilon_0$: Permitivitas vakum (konstanta).
* $r$: Jarak radial dari pusat bola (skalar).
* $a$: Jari-jari bola (skalar).
* $\mathbf{u}_r$: Vektor satuan radial.
* $\mathbf{r}$: Vektor posisi dari pusat bola.

**Untuk $r > a$ (di luar bola):**
$$\mathbf{E} = \frac{q}{4\pi\epsilon_0 r^2} \mathbf{u}_r$$
*Keterangan:* Medan listrik di luar bola, mengikuti hukum kuadrat terbalik.

**Keterangan:**
* Simbol-simbol sama dengan di atas.

**Divergensi $\mathbf{E}$ (untuk $r \le a$):**
$$\nabla \cdot \mathbf{E} = \nabla \cdot \left(\frac{q}{4\pi\epsilon_0 a^3} \mathbf{r}\right) = \frac{q}{4\pi\epsilon_0 a^3} \nabla \cdot (x\mathbf{u}_x + y\mathbf{u}_y + z\mathbf{u}_z) = \frac{q}{4\pi\epsilon_0 a^3} (1+1+1) = \frac{3q}{4\pi\epsilon_0 a^3}$$
*Keterangan:* Divergensi medan listrik di dalam bola adalah konstanta, sesuai dengan kerapatan muatan.

**Keterangan:**
* $x, y, z$: Komponen koordinat Kartesian.
* $\mathbf{u}_x, \mathbf{u}_y, \mathbf{u}_z$: Vektor satuan dalam arah sumbu $x, y, z$.

**Divergensi $\mathbf{E}$ (untuk $r > a$, dalam koordinat bola):**
$$\nabla \cdot \mathbf{E} = \frac{1}{r^2}\frac{\partial}{\partial r}(r^2 E_r) + \frac{1}{r\sin\theta}\frac{\partial}{\partial\theta}(E_\theta \sin\theta) + \frac{1}{r\sin\theta}\frac{\partial E_\phi}{\partial\phi}$$
Karena $E_r = \frac{q}{4\pi\epsilon_0 r^2}$ dan $E_\theta = E_\phi = 0$:
$$\nabla \cdot \mathbf{E} = \frac{1}{r^2}\frac{\partial}{\partial r}\left(r^2 \frac{q}{4\pi\epsilon_0 r^2}\right) = \frac{1}{r^2}\frac{\partial}{\partial r}\left(\frac{q}{4\pi\epsilon_0}\right) = 0$$
*Keterangan:* Divergensi medan listrik di luar bola bermuatan seragam adalah nol, karena tidak ada muatan di luar bola.

**Keterangan:**
* $E_r, E_\theta, E_\phi$: Komponen radial, polar, dan azimut dari medan listrik $\mathbf{E}$ dalam koordinat bola.
* $\theta$: Sudut polar (kolatitude).
* $\phi$: Sudut azimut (longitudinal).
* $\frac{\partial}{\partial r}, \frac{\partial}{\partial\theta}, \frac{\partial}{\partial\phi}$: Operator turunan parsial terhadap $r, \theta, \phi$.

**Curl $\mathbf{E}$ (untuk $r > a$, dalam koordinat bola):**
$$\nabla \times \mathbf{E} = \frac{1}{r\sin\theta}\left[\frac{\partial}{\partial\theta}(E_\phi \sin\theta) - \frac{\partial E_\theta}{\partial\phi}\right]\mathbf{u}_r + \frac{1}{r}\left[\frac{1}{\sin\theta}\frac{\partial E_r}{\partial\phi} - \frac{\partial}{\partial r}(rE_\phi)\right]\mathbf{u}_\theta + \frac{1}{r}\left[\frac{\partial}{\partial r}(rE_\theta) - \frac{\partial E_r}{\partial\theta}\right]\mathbf{u}_\phi$$
Karena $E_r = \frac{q}{4\pi\epsilon_0 r^2}$ dan $E_\theta = E_\phi = 0$:
$$\nabla \times \mathbf{E} = 0$$
*Keterangan:* Curl medan listrik statis adalah nol, menunjukkan medan listrik adalah medan konservatif.

**Keterangan:**
* $\mathbf{u}_\theta$: Vektor satuan polar.
* $\mathbf{u}_\phi$: Vektor satuan azimut.
* Simbol lainnya sama dengan di atas.

**Contoh 2.9: Divergensi dan Curl dari Medan Magnetik**

Untuk medan magnetik $\mathbf{H}$ yang dihasilkan oleh kawat lurus panjang berarus $I$:
$$\mathbf{H} = \frac{I}{2\pi \rho} \mathbf{u}_\phi$$
*Keterangan:* Medan magnetik di sekitar kawat, dengan $\rho$ adalah jarak radial dari kawat dan $\mathbf{u}_\phi$ adalah vektor satuan azimut.

**Keterangan:**
* $\mathbf{H}$: Medan magnetik (vektor).
* $I$: Arus listrik (skalar).
* $\rho$: Jarak radial dari sumbu kawat dalam koordinat silinder (skalar).
* $\mathbf{u}_\phi$: Vektor satuan azimut dalam koordinat silinder.

**Divergensi $\mathbf{H}$ (dalam koordinat silinder):**
$$\nabla \cdot \mathbf{H} = \frac{1}{\rho}\frac{\partial}{\partial\rho}(\rho H_\rho) + \frac{1}{\rho}\frac{\partial H_\phi}{\partial\phi} + \frac{\partial H_z}{\partial z}$$
Karena $H_\phi = \frac{I}{2\pi\rho}$ dan $H_\rho = H_z = 0$:
$$\nabla \cdot \mathbf{H} = \frac{1}{\rho}\frac{\partial}{\partial\rho}(\rho \cdot 0) + \frac{1}{\rho}\frac{\partial}{\partial\phi}\left(\frac{I}{2\pi\rho}\right) + \frac{\partial}{\partial z}(0) = 0$$
*Keterangan:* Divergensi medan magnetik adalah nol, sesuai dengan hukum Gauss untuk magnetisme (tidak ada monopol magnetik).

**Keterangan:**
* $H_\rho, H_\phi, H_z$: Komponen radial, azimut, dan sumbu-z dari medan magnetik $\mathbf{H}$ dalam koordinat silinder.
* $z$: Koordinat sumbu-z.
* $\frac{\partial}{\partial\rho}, \frac{\partial}{\partial\phi}, \frac{\partial}{\partial z}$: Operator turunan parsial terhadap $\rho, \phi, z$.

**Curl $\mathbf{H}$ (dalam koordinat silinder):**
$$\nabla \times \mathbf{H} = \left(\frac{1}{\rho}\frac{\partial H_z}{\partial\phi} - \frac{\partial H_\phi}{\partial z}\right)\mathbf{u}_\rho + \left(\frac{\partial H_\rho}{\partial z} - \frac{\partial H_z}{\partial\rho}\right)\mathbf{u}_\phi + \frac{1}{\rho}\left(\frac{\partial}{\partial\rho}(\rho H_\phi) - \frac{\partial H_\rho}{\partial\phi}\right)\mathbf{u}_z$$
Karena $H_\phi = \frac{I}{2\pi\rho}$ dan $H_\rho = H_z = 0$:
$$\nabla \times \mathbf{H} = 0 \mathbf{u}_\rho + 0 \mathbf{u}_\phi + \frac{1}{\rho}\left(\frac{\partial}{\partial\rho}\left(\rho \frac{I}{2\pi\rho}\right) - 0\right)\mathbf{u}_z = \frac{1}{\rho}\left(\frac{\partial}{\partial\rho}\left(\frac{I}{2\pi}\right)\right)\mathbf{u}_z = 0$$
*Keterangan:* Untuk daerah di luar kawat (di mana tidak ada arus), curl medan magnetik adalah nol. Jika kita mempertimbangkan arus yang mengalir, curl $\mathbf{H}$ akan sebanding dengan kerapatan arus (Hukum Ampere-Maxwell).

**Keterangan:**
* $\mathbf{u}_\rho$: Vektor satuan radial dalam koordinat silinder.
* $\mathbf{u}_z$: Vektor satuan sepanjang sumbu-z dalam koordinat silinder.
* Simbol lainnya sama dengan di atas.

**Contoh 2.11: Curl dan Divergensi Medan Kecepatan Fluida**

Misalkan medan kecepatan fluida $\mathbf{v}$ disekitar silinder diberikan oleh:
$$\mathbf{v}(t, \mathbf{r}) = v_0 \left(1 - \frac{a^2}{r^2}\right) \cos\phi \mathbf{u}_r - v_0 \left(1 + \frac{a^2}{r^2}\right) \sin\phi \mathbf{u}_\phi$$
*Keterangan:* Ini adalah representasi dari aliran fluida yang tidak dapat dimampatkan dan tak-rotasional di sekitar silinder.

**Keterangan:**
* $\mathbf{v}$: Medan kecepatan fluida (vektor).
* $t$: Waktu (skalar).
* $\mathbf{r}$: Vektor posisi.
* $v_0$: Kecepatan aliran bebas (konstanta skalar).
* $a$: Jari-jari silinder (skalar).
* $r$: Jarak radial dari pusat silinder (skalar, dalam konteks ini sama dengan $\rho$ di koordinat silinder).
* $\phi$: Sudut azimut (skalar).
* $\mathbf{u}_r$: Vektor satuan radial (dalam koordinat silinder, sama dengan $\mathbf{u}_\rho$).
* $\mathbf{u}_\phi$: Vektor satuan azimut (dalam koordinat silinder).

**Divergensi $\mathbf{v}$ (dalam koordinat silinder):**
$$\nabla \cdot \mathbf{v} = \frac{1}{\rho}\frac{\partial}{\partial\rho}(\rho v_\rho) + \frac{1}{\rho}\frac{\partial v_\phi}{\partial\phi} + \frac{\partial v_z}{\partial z}$$
Mengganti $v_\rho = v_0 (1 - a^2/r^2) \cos\phi$ dan $v_\phi = -v_0 (1 + a^2/r^2) \sin\phi$, dan $v_z = 0$:
$$\nabla \cdot \mathbf{v} = \frac{1}{\rho}\frac{\partial}{\partial\rho}\left(\rho v_0 \left(1 - \frac{a^2}{\rho^2}\right) \cos\phi\right) + \frac{1}{\rho}\frac{\partial}{\partial\phi}\left(-v_0 \left(1 + \frac{a^2}{\rho^2}\right) \sin\phi\right)$$
Setelah perhitungan turunan parsial, hasilnya adalah:
$$\nabla \cdot \mathbf{v} = 0$$
*Keterangan:* Divergensi nol menunjukkan bahwa fluida tidak dapat dimampatkan (incompressible).

**Keterangan:**
* $v_\rho, v_\phi, v_z$: Komponen radial, azimut, dan sumbu-z dari medan kecepatan $\mathbf{v}$ dalam koordinat silinder.
* Simbol lainnya sama dengan di atas.

**Curl $\mathbf{v}$ (dalam koordinat silinder):**
$$\nabla \times \mathbf{v} = \left(\frac{1}{\rho}\frac{\partial v_z}{\partial\phi} - \frac{\partial v_\phi}{\partial z}\right)\mathbf{u}_\rho + \left(\frac{\partial v_\rho}{\partial z} - \frac{\partial v_z}{\partial\rho}\right)\mathbf{u}_\phi + \frac{1}{\rho}\left(\frac{\partial}{\partial\rho}(\rho v_\phi) - \frac{\partial v_\rho}{\partial\phi}\right)\mathbf{u}_z$$
Mengganti komponen $v_\rho$, $v_\phi$, dan $v_z=0$:
$$\nabla \times \mathbf{v} = 0$$
