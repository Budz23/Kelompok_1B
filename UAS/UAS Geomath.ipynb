# 2.5 Line Integral and Path Independence

**2.5.1 Line Integrals**  
*Line integral* adalah integral yang dihitung sepanjang kurva tertentu di ruang. Dalam konteks vektor, line integral mengukur "kerja" atau kontribusi suatu medan vektor (misalnya medan gaya) sepanjang lintasan (path) tertentu.
Line integral terjadi ketika kita menghitung integral suatu fungsi (skalar atau vektor) sepanjang kurva $ C $. Contohnya, menghitung kerja gaya $ \mathbf{F} $ sepanjang lintasan tertentu.  

```python
import numpy as np
import matplotlib.pyplot as plt

# Definisikan kurva C: r(t) = (t, t), t dari 0 ke 1
def r(t):
    return np.array([t, t])

# Turunan r'(t)
def r_prime(t):
    return np.array([1, 1])

# Medan vektor F(x, y) = (x, y)
def F(x, y):
    return np.array([x, y])

# Fungsi untuk menghitung F(r(t)) . r'(t)
def integrand(t):
    pos = r(t)
    x, y = pos[0], pos[1]
    F_val = F(x, y)
    r_prime_val = r_prime(t)
    return np.dot(F_val, r_prime_val)

# Hitung line integral menggunakan integral numerik
t = np.linspace(0, 1, 1000)
dt = t[1] - t[0]
integral = np.sum(integrand(t) * dt)

# Visualisasi kurva C
t_plot = np.linspace(0, 1, 100)
curve = r(t_plot)
plt.plot(curve[0], curve[1], 'b-', label='Kurva C: r(t) = (t, t)')
plt.scatter([0, 1], [0, 1], color='red', label='Titik awal dan akhir')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Kurva C dan Line Integral')
plt.legend()
plt.grid(True)
plt.show()

# Cetak hasil integral
print(f"Nilai line integral: {integral:.4f}")V
```

$\int_C f(x, y) \, dx = \lim_{\Delta t \to 0} \sum f(x_i, y_i) \cdot \Delta x_i$  
= $\lim_{\Delta t \to 0} \sum f(x(t_i), y(t_i)) \cdot (x(t_i + \Delta t) - x(t_i))$  
= $\lim_{\Delta t \to 0} \sum f(x(t_i), y(t_i)) \cdot \frac{x(t_i + \Delta t) - x(t_i)}{\Delta t} \cdot \Delta t$  
$$
= \int_a^b f(x(t), y(t)) \cdot x'(t) \, dt \tag{1}  
$$  
$\int_C f(x, y) \, dy = \lim_{\Delta t \to 0} \sum f(x_i, y_i) \cdot \Delta y_i$  
= $\lim_{\Delta t \to 0} \sum f(x(t_i), y(t_i)) \cdot (y(t_i + \Delta t) - y(t_i))$  
= $\lim_{\Delta t \to 0} \sum f(x(t_i), y(t_i)) \cdot \frac{y(t_i + \Delta t) - y(t_i)}{\Delta t} \cdot \Delta t$  
$$
= \int_a^b f(x(t), y(t)) \cdot y'(t) \, dt \tag{2} 
$$
$\int_C f(x, y) \, dl = \lim_{\Delta t \to 0} \sum f(x_i, y_i) \cdot \sqrt{(\Delta x_i)^2 + (\Delta y_i)^2}$  
= $\lim_{\Delta t \to 0} \sum f(x(t_i), y(t_i)) \cdot \sqrt{\left(\frac{\Delta x_i}{\Delta t}\right)^2 + \left(\frac{\Delta y_i}{\Delta t}\right)^2} \cdot \Delta t$  
$$
= \int_a^b f(x(t), y(t)) \cdot \sqrt{(x'(t))^2 + (y'(t))^2} \, dt \tag{3} 
$$  
## (Example 2.5.1) Line Integrals on a Plane
Evaluate the following three integrals
$$
(a) \int_C x^2 y \, dx\tag{E.2.5.1.1}

$$
(b) \int_C x^2 y \, dy\tag{E.2.5.1.2}

$$
(c) \int_C x^2 y \, dl\tag{E.2.5.1.3}

on the quarter-circle  parameterized by $(2 \cos t u_x + 2 \sin t u_y)$ for $(0 \leq t \leq \pi/2)$.

```python
import sympy as sp

# Definisikan variabel dan parameter
t = sp.Symbol('t')
x = 2 * sp.cos(t)
y = 2 * sp.sin(t)
f = x**2 * y
dx_dt = sp.diff(x, t)

# Batas integrasi
a, b = 0, sp.pi/2

# Integral
integral = sp.integrate(f * dx_dt, (t, a, b))

print("Hasil integral (a):", integral)
```

***2.5.1.2 Integral Garis Fungsi Skalar***

Definisi: Untuk fungsi skalar $f(x, y, z)$ sepanjang kurva $C$ yang diparameterisasi oleh $x(t), y(t), z(t)$, $a \leq t \leq b$:    

$\int_C$ $f(x, y, z)$ , $dl$ = $\int_a^b f(x(t), y(t), z(t)) \sqrt{(x'(t))^2 + (y'(t))^2 + (z'(t))^2} , dt$  
$$ 
\int_C f(x, y, z) \, dx = \int_a^b f(x(t), y(t), z(t)) \, x'(t) \, dt  \tag{4}  
$$
$$
\int_C f(x, y, z) \, dy = \int_a^b f(x(t), y(t), z(t)) \, y'(t) \, dt \tag{5}  
$$
$$
\int_C f(x, y, z) \, dz = \int_a^b f(x(t), y(t), z(t)) \, z'(t) \, dt \tag{6}  
$$
$$
\int_C f(x, y, z) \, dl = \int_a^b f(x(t), y(t), z(t)) \sqrt{(x'(t))^2 + (y'(t))^2 + (z'(t))^2} \, dt \tag{7}  
$$


## Example (2.5.2) Line Integral of a Scale Function in 3D Space 
Evaluasi integral berikut berdasarkan panjang busur $l$:  
$$
\int_C 8xyz \, dl\tag{E.2.5.2}
$$

```python
import numpy as np

# Definisikan parameterisasi kurva r(u) = (u*cos(u), u*sin(u), 0)
def r(u):
    return np.array([u * np.cos(u), u * np.sin(u), 0])

# Turunan r'(u)
def r_prime(u):
    return np.array([np.cos(u) - u * np.sin(u), np.sin(u) + u * np.cos(u), 0])

# Fungsi skalar 8xyz
def f(x, y, z):
    return 8 * x * y * z

# Elemen panjang busur dl = |r'(u)| du
def dl(u):
    return np.sqrt(np.sum(r_prime(u) ** 2))

# Fungsi integrand: 8xyz * dl
def integrand(u):
    pos = r(u)
    x, y, z = pos[0], pos[1], pos[2]
    return f(x, y, z) * dl(u)

# Hitung integral numerik
u = np.linspace(0, np.pi, 1000)
du = u[1] - u[0]
integral = np.trapz(integrand(u), u)

# Cetak hasil
print(f"Nilai integral numerik: {integral:.4f}")
```

***2.5.1.3 Line Integral of a Vector Function In Space***  
Adalah integral dari suatu medan vektor $ \mathbf{F}(x, y, z) = F_x \mathbf{u}_x + F_y \mathbf{u}_y + F_z \mathbf{u}_z $ sepanjang kurva halus $ C $. Integral ini menghitung "kerja" atau kontribusi medan vektor sepanjang lintasan $ C $, dihitung dengan mengintegralkan komponen medan sepanjang vektor tangen diferensial kurva. Kurva $C$ diparameterisasi oleh:  
$ \mathbf{r}(t) = x(t) \mathbf{u}_x + y(t) \mathbf{u}_y + z(t) \mathbf{u}_z $, $ t $ dari $ a $ hingga $ b $.

$ r(t) = x(t) u_x + y(t) u_y + z(t) u_z $, $ t $ dari $ a $ hingga $ b $.  Vektor tangen didefinisikan sebagai turunan $r(t)$ terhadap $t$:
$$
\frac{dr}{dt} = \frac{dx}{dt} u_x + \frac{dy}{dt} u_y + \frac{dz}{dt} u_z \tag{8}
$$
$$
dl = \frac{dr}{dt} dt \tag{9}
$$
$ \mathbf{F}(x, y, z) = F_x \mathbf{u}_x + F_y \mathbf{u}_y + F_z \mathbf{u}_z $ didefinisikan sebagai:
$$
\int_C F \cdot dl \tag{10}
$$
$$
F \cdot dl = F \cdot \frac{dr}{dt} dt 
$$
$$
F \cdot \frac{dr}{dt} = (F_x u_x + F_y u_y + F_z u_z) \cdot \left( \frac{dx}{dt} u_x + \frac{dy}{dt} u_y + \frac{dz}{dt} u_z \right) 
$$
$$
= F_x \frac{dx}{dt} + F_y \frac{dy}{dt} + F_z \frac{dz}{dt} \tag{11}
$$
$$
F_x(x(t), y(t), z(t)) \tag{12}
$$
Maka, $ \int_C F \cdot dl$
$$
= \int_a^b \left[ F_x(x(t), y(t), z(t)) \frac{dx}{dt} + F_y(x(t), y(t), z(t)) \frac{dy}{dt} + F_z(x(t), y(t), z(t)) \frac{dz}{dt} \right] dt \tag{13}
$$

**2.5.2 Path Independence**   
*Path independence* mengacu pada sifat line integral dari suatu medan vektor $ G(x, y, z) = G_x u_x + G_y u_y + G_z u_z $ yang nilainya hanya bergantung pada titik awal dan akhir, bukan lintasan yang ditempuh. Medan vektor disebut konservatif jika dapat ditulis sebagai gradien dari fungsi skalar $ f(x, y, z) $, yang disebut fungsi potensial. Kriteria konservatif adalah curl medan nol ($ \nabla \times G = 0 $) di wilayah sederhana yang terhubung, dan turunan parsialnya kontinu.

$ r(t) = x(t) u_x + y(t) u_y + z(t) u_z $, $ t $ dari $ a $ hingga $ b $. 
$$ 
\frac{dr}{dt} = \frac{dx}{dt} u_x + \frac{dy}{dt} u_y + \frac{dz}{dt} u_z
$$
$$
dl = \frac{dr}{dt} dt
$$
$$ 
\int_C G \cdot dl \tag{14}
$$
$$
G \cdot dl = G \cdot \frac{dr}{dt} dt
$$
$$
G \cdot \frac{dr}{dt} = (G_x u_x + G_y u_y + G_z u_z) \cdot \left( \frac{dx}{dt} u_x + \frac{dy}{dt} u_y + \frac{dz}{dt} u_z \right)
$$
$$
= G_x \frac{dx}{dt} + G_y \frac{dy}{dt} + G_z \frac{dz}{dt} \tag{15}
$$ 
$$
G_x(x(t), y(t), z(t)), dst.
$$
Jika $ G $ konservatif, maka $ G = \nabla f $.  
$$
\nabla f = \frac{\partial f}{\partial x} u_x + \frac{\partial f}{\partial y} u_y + \frac{\partial f}{\partial z} u_z \tag{16}
$$
$$
G_x = \frac{\partial f}{\partial x} , G_y = \frac{\partial f}{\partial y} , G_z = \frac{\partial f}{\partial z} \tag{17}
$$
Substitusi ke integral:
$$
\int_C G \cdot dl = \int_a^b \left[ \frac{\partial f}{\partial x} \frac{dx}{dt} + \frac{\partial f}{\partial y} \frac{dy}{dt} + \frac{\partial f}{\partial z} \frac{dz}{dt} \right] dt \tag{18}
$$
Gunakan aturan rantai: $ \frac{d}{dt} f(x(t), y(t), z(t)) $
$$ 
= \frac{\partial f}{\partial x} \frac{dx}{dt} + \frac{\partial f}{\partial y} \frac{dy}{dt} + \frac{\partial f}{\partial z} \frac{dz}{dt} \tag{19}
$$
Jadi,
$$
\int_a^b \frac{d}{dt} f(x(t), y(t), z(t)) dt  
$$
$$= f(r(b)) - f(r(a)) \tag{20}
$$
Hasil akhir:
$$
\int_C G \cdot dl = f(B) - f(A) \quad  dengan \quad A = r(a) \quad dan \quad  B = r(b)
$$

## Example (2.5.3) Path Independence of Line Integrals  
Tentukan kemandirian lintasan (path independence) dari integral garis berikut berdasarkan apakah fungsi vektor integran konservatif:

$$
\text {(a)}
\int_C G(x, y) \cdot dl = \int_C [G_x(x, y) u_x + G_y(x, y) u_y] \cdot dl = \int_C \frac{-y u_x + x u_y}{x^2 + y^2} \cdot dl
$$

Fungsi vektor $ G(x, y) $ memenuhi $ \frac{\partial G_y}{\partial x} = \frac{\partial G_x}{\partial y} $ untuk medan vektor konservatif 2D:  
$$
\frac{\partial G_y}{\partial x} = \frac{(x^2 + y^2) - 2x}{ (x^2 + y^2)^2} = \frac{y^2 - x^2}{ (x^2 + y^2)^2} = -\frac{(x^2 + y^2) + y^2}{ (x^2 + y^2)^2} \frac{\partial G_x}{\partial y}
$$

Namun, diketahui bahwa integral garis ini tidak path-independent karena adanya singularitas di titik (0, 0), yang merupakan lubang dalam domain terbuka $ D $ yang mengandung lintasan integrasi.

$$
\text {(b)}
\int_C G(x, y) \cdot dl = \int_C [G_x(x, y) u_x + G_y(x, y) u_y] \cdot dl = \int_C \frac{-y u_x + x u_y}{x^2} \cdot dl 
$$

Fungsi vektor $ G(x, y) $ memenuhi kriteria $ \frac{\partial G_y}{\partial x} = \frac{\partial G_x}{\partial y} $ untuk medan vektor konservatif 2D:  
$$
 \frac{\partial G_y}{\partial x} = -\frac{1}{x^2} = \frac{\partial G_x}{\partial y} 
$$

melalui dua wilayah sederhana yang terhubung, dipisahkan oleh $ x = 0 $. Fungsi potensial skalar $ f(x, y) = \frac{y}{x} + C $ dengan $ \frac{\partial f}{\partial x} = -\frac{y}{x^2} $ dan $ \frac{\partial f}{\partial y} = \frac{1}{x} $.


```python
import numpy as np

# Definisikan fungsi vektor G(x, y)
def G(x, y):
    return np.array([-y / x**2, 1 / x])

# Parameterisasi lintasan C1: r(t) = t*u + (1-t)*u, 0 <= t <= 1 (dari (0,0) ke (1,1))
def r1(t):
    return np.array([t, t])  # Misal u = (1,1) untuk simplifikasi

def r1_prime(t):
    return np.array([1, 1])

# Parameterisasi lintasan C2: r(t) = (2-t)*u, 0 <= t <= 1 (dari (2,2) ke (1,1))
def r2(t):
    return np.array([2 - t, 2 - t])

def r2_prime(t):
    return np.array([-1, -1])

# Fungsi integrand untuk G . dr/dt
def integrand1(t):
    pos = r1(t)
    x, y = pos[0], pos[1]
    G_val = G(x, y)
    r_prime_val = r1_prime(t)
    return np.dot(G_val, r_prime_val)

def integrand2(t):
    pos = r2(t)
    x, y = pos[0], pos[1]
    G_val = G(x, y)
    r_prime_val = r2_prime(t)
    return np.dot(G_val, r_prime_val)

# Hitung integral numerik menggunakan trapezoidal rule
t = np.linspace(0, 1, 1000)
dt = t[1] - t[0]
integral1 = np.trapz(integrand1(t), t)
integral2 = np.trapz(integrand2(t), t)
integral_total = integral1 + integral2

# Cetak hasil
print(f"Nilai integral C1: {integral1:.4f}")
print(f"Nilai integral C2: {integral2:.4f}")
print(f"Nilai integral total (C1 + C2): {integral_total:.4f}")
```

# 2.6 Double Integrals  
adalah integral ganda dari fungsi $ f(x, y) $ atau $ f(\rho, \phi) $ over wilayah $ R $ di bidang $ xy $ atau $ \rho\phi $, yang menghitung volume 3D jika $ f(x, y) $ adalah tinggi, atau luas jika $ f(x, y) = 1 $. Dalam koordinat Cartesian, integral dihitung sebagai iterasi ganda tergantung bentuk $ R $ (Tipe I atau Tipe II). Dalam koordinat polar, digunakan untuk fungsi bergantung pada $ x^2 + y^2 $. Double Integral terjadi ketika kita mengintegralkan fungsi dua variabel over wilayah dua dimensi $ R $

Wilayah $ R $ dibagi menjadi elemen kecil $ dA = dy dx $ atau $ dx dy $.  
Untuk Tipe I: 
$$
R = {(x, y) \mid a \leq x \leq b, g_1(x) \leq y \leq g_2(x)} \tag{21}
$$ 
integrasi dalam terhadap $ y $, lalu $ x $,
$$
\iint_R f(x, y) dA = \int_a^b \int_{g_1(x)}^{g_2(x)} f(x, y) dy dx \tag{22}
$$   
$ dA = dy dx $ karena $ y $ bervariasi antara $ g_1(x) $ dan $ g_2(x) $ untuk $ x $ tetap.  
$ \int_{g_1(x)}^{g_2(x)} f(x, y) dy $ adalah kontribusi $ y $ pada $ x $ tetap.  
$ \int_a^b (...) dx $ menjumlahkan semua $ x $ dari $ a $ hingga $ b $.  
Untuk Tipe II.
$$
R = {(x, y) \mid c \leq y \leq d, h_1(y) \leq x \leq h_2(y)} \tag{23}
$$
integrasi dalam terhadap $ x $, lalu $ y $.
$$
\iint_R f(x, y) dA = \int_c^d \int_{h_1(y)}^{h_2(y)} f(x, y) dx dy  \tag{24}
$$  
$ dA = dx dy $ karena $ x $ bervariasi antara $ h_1(y) $ dan $ h_2(y) $ untuk $ y $ tetap.  
$ \int_{h_1(y)}^{h_2(y)} f(x, y) dx $ adalah kontribusi $ x $ pada $ y $ tetap.  
$ \int_c^d (...) dy $ menjumlahkan semua $ y $ dari $ c $ hingga $ d $.  
Konversi ke Koordinat Polar.  
$$
x = \rho \cos \phi, y = \rho \sin \phi, dA = \rho d\rho d\phi
$$ 
(faktor $ \rho $ dari Jacobian).  
Untuk $ R = {(\rho, \phi) \mid \alpha \leq \phi \leq \beta, r_1(\phi) \leq \rho \leq r_2(\phi)} $.
Integrasi dalam terhadap $ \rho $, lalu $ \phi $,
$$
\iint_R f(\rho, \phi) dA = \int_{\alpha}^{\beta} \int_{r_1(\phi)}^{r_2(\phi)} f(\rho, \phi) \rho d\rho d\phi \tag{25}
$$

***2.6.1 Center of Mass and Moments of Laminas with Variable Density***  
mengacu pada perhitungan massa $ m = \iint_R \rho(x, y) dA $ dan koordinat pusat massa $ (\bar{x}, \bar{y}) = \left( \frac{M_y}{m}, \frac{M_x}{m} \right) $ dari lamina dengan densitas variabel $ \rho(x, y) $ yang kontinu di wilayah $ R $. Momen pertama $ M_x $ dan $ M_y $ dihitung menggunakan integral ganda, dan konversi ke koordinat polar memungkinkan perhitungan yang lebih mudah untuk fungsi bergantung pada $ x^2 + y^2 $.

Koordinat Pusat Massa $ (\bar{x}, \bar{y}) $.  
Pusat massa adalah titik keseimbangan, di mana momen dibagi massa.  
$$ 
\bar{x} = \frac{M_y}{m} = \frac{\iint_R x \rho(x, y) dA}{\iint_R \rho(x, y) dA}, \tag{26}
$$ 
rata-rata $x$ tertimbang oleh densitas.
$$ 
\bar{y} = \frac{M_x}{m} = \frac{\iint_R y \rho(x, y) dA}{\iint_R \rho(x, y) dA}, \tag{27}
$$
rata-rata  $y$ tertimbang oleh densitas.   
Konversi ke Koordinat Polar.  
$$
 x = \rho \cos \phi, y = \rho \sin \phi, dA = \rho d\rho d\phi \tag{28}
$$  
$$
m = \int_{\alpha}^{\beta} \int_{r_1(\phi)}^{r_2(\phi)} \rho(\rho \cos \phi, \rho \sin \phi) \rho d\rho d\phi \tag{29}
$$  
$$

# 2.7 Green's Theorerm
menyatakan hubungan antara integral garis sepanjang kurva tertutup sederhana $ C $ (arah counterclockwise) dan integral ganda over wilayah $ R $ yang dibatasi oleh $ C $. Jika $ P $ dan $ Q $ adalah fungsi dengan turunan parsial kontinu, teorema ini menyatakan bahwa 
$$ 
\oint_C P dx + Q dy = \iint_R \left( \frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y} \right) dA \tag{32}
$$
Green's Theorerm terjadi ketika kita menghubungkan integral garis sepanjang kurva tertutup dengan integral ganda di wilayah yang dibatasinya, dengan asumsi kondisi kontinuitas turunan.

Integral garis: 
$$ 
\oint_C P dx + Q dy 
$$
dihitung sepanjang kurva tertutup $ C $ dengan arah counterclockwise.  
Integral ganda: 
$$ 
\iint_R \left( \frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y} \right) dA 
$$ 
dihitung over wilayah $ R $ yang dibatasi $ C $, di mana $ dA $ adalah elemen luas.  
Dekomposisi Wilayah $ R $.  
Misalkan $ R $ dapat dibagi menjadi tipe I: 
$$ 
a \leq x \leq b ,  g_1(x) \leq y \leq g_2(x) , atau \quad tipe II: c \leq y \leq d ,  h_1(y) \leq x \leq h_2(y)
$$  
Fokus pada tipe I untuk penurunan awal.  
Kontribusi 
$$ 
\frac{\partial P}{\partial y} dA 
$$  
$$
-\iint_R \frac{\partial P}{\partial y} dA = -\int_a^b \int_{g_1(x)}^{g_2(x)} \frac{\partial P}{\partial y} dy dx
$$
Terapkan teorema dasar kalkulus: 
$$ 
\int_{g_1(x)}^{g_2(x)} \frac{\partial P}{\partial y} dy = P(x, g_2(x)) - P(x, g_1(x))
$$
Jadi, 
$$ 
-\int_a^b [P(x, g_2(x)) - P(x, g_1(x))] dx 
$$
Dekomposisi Integral Garis untuk $ P dx $.  
Kurva $ C $ terdiri dari $ C_1 $ (dari $ (a, g_1(a)) $ ke $ (b, g_2(b)) $) dan $ C_2 $ hingga $ C_4 $ (sekitar batas).  
$$
\int_{C_1} P dx = \int_a^b P(x, g_2(x)) dx, \int_{C_3} P dx = -\int_a^b P(x, g_1(x)) dx 
$$ 
(arah berlawanan).  
Total   
$$ 
\int_C P dx = \int_{C_1} P dx + \int_{C_3} P dx + \int_{C_2} P dx + \int_{C_4} P dx 
$$  
Kontribusi $ \frac{\partial Q}{\partial x} dA $.  
$$ 
\iint_R \frac{\partial Q}{\partial x} dA = \int_c^d \int_{h_1(y)}^{h_2(y)} \frac{\partial Q}{\partial x} dx dy
$$
Terapkan teorema dasar: 
$$ 
\int_{h_1(y)}^{h_2(y)} \frac{\partial Q}{\partial x} dx = Q(h_2(y), y) - Q(h_1(y), y) 
$$
Jadi, 
$$ 
\int_c^d [Q(h_2(y), y) - Q(h_1(y), y)] dy
$$
Dekomposisi Integral Garis untuk $ Q dy $.  
$$ 
\int_{C_2} Q dy = \int_c^d Q(h_2(y), y) dy ,  \int_{C_4} Q dy = -\int_c^d Q(h_1(y), y) dy
$$
Total $ \int_C Q dy = \int_{C_2} Q dy + \int_{C_4} Q dy $.

# 2.8 Surface Integral  
adalah integral ganda yang dihitung di atas permukaan $ S $ dalam ruang tiga dimensi. Ada dua jenis utama:   
(1) Surface Integral of a Scalar Function $ \iint_S f(x, y, z) dS $, yang menghitung jumlah tertimbang dari fungsi skalar $ f $ di seluruh permukaan, dan   
(2) Surface Integral of a Vector Field $ \iint_S F \cdot dS $, yang menghitung fluks vektor $ F $ melalui permukaan.  
Untuk permukaan berparameter $ r(u, v) $, elemen permukaan $ dS $ ditentukan oleh vektor normal dan diferensial area.  
Surface Integral terjadi saat mengintegralkan fungsi skalar atau vektor di atas permukaan di ruang 3D, sering digunakan untuk menghitung luas, fluks, atau massa permukaan.

Skalar: 
$$ 
\iint_S f(x, y, z) dS = \iint_D f(r(u, v)) \left| \frac{\partial r}{\partial u} \times \frac{\partial r}{\partial v} \right| du dv
$$

Vektor: 
$$ 
\iint_S F \cdot dS = \iint_D F(r(u, v)) \cdot \left( \frac{\partial r}{\partial u} \times \frac{\partial r}{\partial v} \right) du dv 
$$

Permukaan $ S $ didefinisikan oleh parameterisasi:

$$
r(u, v) = x(u, v) u_x + y(u, v) u_y + z(u, v) u_z ,  
$$

di mana: $ (u, v) \in D $ (domain 2D).  

Elemen permukaan $ dS $ dihitung berdasarkan vektor normal dan diferensial area.  
Vektor tangen:
$$
\frac{\partial r}{\partial u} = \frac{\partial x}{\partial u} u_x + \frac{\partial y}{\partial u} u_y + \frac{\partial z}{\partial u} u_z 
$$  
Vektor tangen $ \frac{\partial r}{\partial v} = \frac{\partial x}{\partial v} u_x + \frac{\partial y}{\partial v} u_y + \frac{\partial z}{\partial v} u_z $.  
Vektor normal $ N = \frac{\partial r}{\partial u} \times \frac{\partial r}{\partial v} $, yang merupakan hasil kali vektor tangen.
Elemen Permukaan $ dS $.  
Besaran $ \left| \frac{\partial r}{\partial u} \times \frac{\partial r}{\partial v} \right| du dv $ mewakili elemen area diferensial pada permukaan.  
Ini dihasilkan dari norm vektor silang, yang mengukur luas paralelogram yang dibentuk oleh vektor tangen.
Surface Integral of a Scalar Function.  
$ \iint_S f(x, y, z) dS $ menghitung jumlah $ f $ tertimbang oleh area permukaan.  
Substitusi parameter: $ \iint_S f(x, y, z) dS = \iint_D f(r(u, v)) \left| \frac{\partial r}{\partial u} \times \frac{\partial r}{\partial v} \right| du dv $.  
Integrasi dilakukan di domain $ D $ dengan $ dS $ sebagai elemen area.  
Surface Integral of a Vector Field.  
$ \iint_S F \cdot dS $ menghitung fluks vektor $ F $ melalui permukaan, dengan $ dS = N dS $ (vektor normal kali elemen area).  
Substitusi parameter: $ \iint_S F \cdot dS = \iint_D F(r(u, v)) \cdot \left( \frac{\partial r}{\partial u} \times \frac{\partial r}{\partial v} \right) du dv $.  
Titik kali $ F \cdot N $ mencerminkan komponen vektor normal ke permukaan.  
Kasus Permukaan Explisit $ z = g(x, y) $.  
Jika $ S $ didefinisikan sebagai $ z = g(x, y) $ di $ D $, maka $ r(x, y) = x u_x + y u_y + g(x, y) u_z $.  
$ \frac{\partial r}{\partial x} = u_x + \frac{\partial g}{\partial x} u_z $, $ \frac{\partial r}{\partial y} = u_y + \frac{\partial g}{\partial y} u_z $.  
Vektor normal: $ N = \left( -\frac{\partial g}{\partial x}, -\frac{\partial g}{\partial y}, 1 \right) $, dan $ dS = \sqrt{1 + \left( \frac{\partial g}{\partial x} \right)^2 + \left( \frac{\partial g}{\partial y} \right)^2} dx dy $.  
Jadi, $ \iint_S f dS = \iint_D f(x, y, g(x, y)) \sqrt{1 + \left( \frac{\partial g}{\partial x} \right)^2 + \left( \frac{\partial g}{\partial y} \right)^2} dx dy $.
