### 1.1.1 GEOMETRY WITH VECTORS

Skalar adalah besaran yang hanya memiliki magnitudo (panjang), contohnya: massa, panjang, suhu.

Vektor adalah besaran yang memiliki magnitudo dan arah, direpresentasikan:
Secara grafis: panah berarah.
Secara aljabar: pasangan berurutan [ax ay]ᵀ (untuk 2D) atau [ax ay az]ᵀ (untuk 3D).

Contoh:
Vektor a = [ax ay az]ᵀ menyatakan arah dari titik asal (0,0,0) ke (ax, ay, az), dengan panjang:
$$
||\mathbf{a}|| = \sqrt{a_x^2 + a_y^2 + a_z^2}
$$


Vektor sering ditulis sebagai: 
$\mathbf{a} = a_x \mathbf{u}_x + a_y \mathbf{u}_y + a_z \mathbf{u}_z$




Dengan:
$$
\mathbf{u}_x = \begin{bmatrix}1 \\ 0 \\ 0\end{bmatrix},\quad
\mathbf{u}_y = \begin{bmatrix}0 \\ 1 \\ 0\end{bmatrix},\quad
\mathbf{u}_z = \begin{bmatrix}0 \\ 0 \\ 1\end{bmatrix}
$$


Misal:
$$
\mathbf{a} =
\begin{bmatrix}
2 \\
1 \\
1
\end{bmatrix}, \quad
\mathbf{b} =
\begin{bmatrix}
1 \\
0 \\
-1
\end{bmatrix}
$$


Penjumlahan:
$$
\mathbf{a} + \mathbf{b} = 
\begin{bmatrix}
3 \\
1 \\
0
\end{bmatrix}
$$


Skalar x Vektor:
$$
2\mathbf{a} = 
\begin{bmatrix}
4 \\
2 \\
2
\end{bmatrix}
$$


Panjang Vektor:
$$
||\mathbf{a}|| = \sqrt{2^2 + 1^2 + 1^2} = \sqrt{6}
$$


Contoh: 
Tuliskan vektor a = [3, -2, 5]ᵀ dalam bentuk kombinasi unit vektor.

```python
import numpy as np

# Vektor a
a = np.array([3, -2, 5])

# Unit vectors
ux = np.array([1, 0, 0])
uy = np.array([0, 1, 0])
uz = np.array([0, 0, 1])

# Representasi kombinasi linear
a_vector = a[0]*ux + a[1]*uy + a[2]*uz

# Tampilkan hasil
print(f"Vektor a sebagai kombinasi unit vektor:")
print(f"{a[0]}*ux + ({a[1]})*uy + {a[2]}*uz")
print(f"Hasil vektor: {a_vector}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Vektor a
a = np.array([3, -2, 5])

# Unit vectors
ux = np.array([1, 0, 0])
uy = np.array([0, 1, 0])
uz = np.array([0, 0, 1])

# Basis kombinasi
ax = a[0] * ux
ay = a[1] * uy
az = a[2] * uz

# Setup plot
fig = plt.figure()
ax3d = fig.add_subplot(111, projection='3d')
ax3d.set_xlim([0, 4])
ax3d.set_ylim([-3, 2])
ax3d.set_zlim([0, 6])

# Plot unit vector components
ax3d.quiver(0, 0, 0, ax[0], ax[1], ax[2], color='r', label='a_x * u_x')
ax3d.quiver(ax[0], 0, 0, 0, ay[1], 0, color='g', label='a_y * u_y')
ax3d.quiver(ax[0], ay[1], 0, 0, 0, az[2], color='b', label='a_z * u_z')

# Plot resultant vector a
ax3d.quiver(0, 0, 0, a[0], a[1], a[2], color='k', linewidth=2, label='a = [3, -2, 5]')

# Labeling
ax3d.set_xlabel('X')
ax3d.set_ylabel('Y')
ax3d.set_zlabel('Z')
ax3d.set_title('Visualisasi Vektor a sebagai Kombinasi Unit Vektor')
ax3d.legend()

plt.tight_layout()
plt.show()
```

### 1.1.2 DOT PRODUCT

Dot product, juga dikenal sebagai produk skalar atau inner product, adalah operasi aljabar antara dua vektor yang menghasilkan skalar. 

Misalnya, jika dua vektor 𝑎 dan 𝑏 didefinisikan sebagai:

$$
\mathbf{a} = a_x \mathbf{u}_x + a_y \mathbf{u}_y + a_z \mathbf{u}_z, \quad 
\mathbf{b} = b_x \mathbf{u}_x + b_y \mathbf{u}_y + b_z \mathbf{u}_z
$$



Maka dot product didefinisikan sebagai:
$$
\mathbf{a}^T \mathbf{b} = a_x b_x + a_y b_y + a_z b_z = \|\mathbf{a}\| \cdot \|\mathbf{b}\| \cdot \cos(\theta)
$$


dengan 𝜃 adalah sudut antara kedua vektor.

Rumus-Rumus Penting dalam Dot Product: 

a. Norma atau Panjang Vektor
$$
\|\mathbf{a}\| = \sqrt{\mathbf{a}^T \mathbf{a}} = \sqrt{a_x^2 + a_y^2 + a_z^2}
$$


b. Sudut antar Dua Vektor
$$
\cos(\theta) = \frac{\mathbf{a}^T \mathbf{b}}{\|\mathbf{a}\| \cdot \|\mathbf{b}\|}
$$


c. Proyeksi Vektor 𝑎 ke 𝑏
$$
\text{comp}_{\mathbf{b}}(\mathbf{a}) = \frac{\mathbf{a}^T \mathbf{b}}{\|\mathbf{b}\|}
$$
$$
\text{proj}_{\mathbf{b}}(\mathbf{a}) = \left( \frac{\mathbf{a}^T \mathbf{b}}{||\mathbf{b}||^2} \right) \cdot \mathbf{b}
$$



d. Jarak Titik ke Bidang
Jika titik $\mathbf{p}_1 = \begin{bmatrix} x_1 \\ y_1 \\ z_1 \end{bmatrix}$ dan bidang dengan persamaan ax+by+cz+d=0, maka jaraknya:
$$
\text{Distance} = \frac{ \left| ax_1 + by_1 + cz_1 + d \right| }{ \sqrt{a^2 + b^2 + c^2} }
$$






e. Cauchy–Schwarz Inequality
$$
|\mathbf{a}^T \mathbf{b}| \leq \|\mathbf{a}\| \cdot \|\mathbf{b}\|
$$


f. Triangle Inequality
$$
\|\mathbf{a} + \mathbf{b}\| \leq \|\mathbf{a}\| + \|\mathbf{b}\|
$$



Interpretasi Geometris Dot Product:

Jika $\mathbf{a}^T \mathbf{b}$ = 0, maka $ a \perp b $ (saling tegak lurus).

Semakin kecil sudut 𝜃, semakin besar nilai dot product.

Jika $\mathbf{a}  dan  \mathbf{b}$ paralel searah, maka cos(𝜃)=1

Jika berlawanan arah, cos(𝜃)=−1

Contoh: Mencari Sudut antara Dua Vektor

Diberikan:
$$
\mathbf{a}^T =
\begin{bmatrix}
2 \\
1 \\
3
\end{bmatrix}, \quad
\mathbf{b}^T =
\begin{bmatrix}
1 \\
1 \\
2
\end{bmatrix}
$$




```python
import numpy as np
import math

# Diberikan dua vektor a dan b
a = np.array([2, 1, 3])
b = np.array([1, 1, 2])

# Hitung dot product (a · b)
dot_product = np.dot(a, b)

# Hitung norma (magnitudo) masing-masing vektor
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)

# Hitung cos(θ)
cos_theta = dot_product / (norm_a * norm_b)

# Hitung sudut dalam radian, lalu ubah ke derajat
theta_rad = np.arccos(cos_theta)
theta_deg = np.degrees(theta_rad)

# Tampilkan hasil
print(f"Dot product a · b = {dot_product}")
print(f"Norma ||a|| = {norm_a:.4f}")
print(f"Norma ||b|| = {norm_b:.4f}")
print(f"cos(θ) = {cos_theta:.4f}")
print(f"Sudut antara a dan b = {theta_deg:.2f} derajat")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Vektor a dan b
a = np.array([2, 1, 3])
b = np.array([1, 1, 2])

# Dot product dan norm
dot_product = np.dot(a, b)
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)
cos_theta = dot_product / (norm_a * norm_b)
theta_deg = np.degrees(np.arccos(cos_theta))

# Cetak info
print(f"Dot product a · b = {dot_product}")
print(f"Norma ||a|| = {norm_a:.4f}")
print(f"Norma ||b|| = {norm_b:.4f}")
print(f"cos(θ) = {cos_theta:.4f}")
print(f"Sudut antara a dan b = {theta_deg:.2f} derajat")

# Plot 3D
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Vektor dari titik asal (0,0,0)
ax.quiver(0, 0, 0, a[0], a[1], a[2], color='r', label='a = [2,1,3]', arrow_length_ratio=0.1)
ax.quiver(0, 0, 0, b[0], b[1], b[2], color='b', label='b = [1,1,2]', arrow_length_ratio=0.1)

# Titik label untuk sudut
mid_point = (a / norm_a + b / norm_b)
ax.text(mid_point[0], mid_point[1], mid_point[2], f"{theta_deg:.1f}°", color='black')

# Label sumbu dan set batas
ax.set_xlim(0, 3)
ax.set_ylim(0, 3)
ax.set_zlim(0, 4)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Visualisasi 3D: Sudut antara vektor a dan b')
ax.legend()

# Penting! Ini untuk menampilkan plot
plt.show()
```

### 1.1.3 CROSS PRODUCT

Cross product atau perkalian silang dari dua vektor dalam ruang tiga dimensi adalah suatu operasi vektor yang menghasilkan vektor baru yang tegak lurus (orthogonal) terhadap kedua vektor asal. Cross product hanya terdefinisi dalam ruang tiga dimensi dan dua dimensi tertentu (yang diangkat ke 3D secara implisit).

Jika terdapat dua vektor:
$$
\mathbf{a} = \begin{bmatrix} a_x \\ a_y \\ a_z \end{bmatrix},\quad
\mathbf{b} = \begin{bmatrix} b_x \\ b_y \\ b_z \end{bmatrix}
$$




Maka cross product ditulis sebagai:
$$
\mathbf{a} \times \mathbf{b} = 
\begin{vmatrix}
\mathbf{u}_x & \mathbf{u}_y & \mathbf{u}_z \\
a_x & a_y & a_z \\
b_x & b_y & b_z
\end{vmatrix}
$$


Atau secara komponen:
$$
\mathbf{a} \times \mathbf{b} =
\left[ a_y b_z - a_z b_y,\ a_z b_x - a_x b_z,\ a_x b_y - a_y b_x \right]^T
$$



Ciri-ciri dan Sifat Penting:

a. Magnitude (panjang) dari hasil cross product:
$$
\| \mathbf{a} \times \mathbf{b} \| = \| \mathbf{a} \| \| \mathbf{b} \| \sin(\theta)
$$
di mana θ adalah sudut antara 𝑎 dan 𝑏.

b. Arah ditentukan oleh aturan tangan kanan (right-hand rule).

c. Cross product dua vektor sejajar = 0, karena sudut antara mereka adalah 0° atau 180°, dan sin(0)=sin(180°)=0


Contoh: Luas segitiga dengan vektor sisi

Diberikan:
$$
\mathbf{a} = \begin{bmatrix} 2 \\ 1 \\ 0 \end{bmatrix},\quad
\mathbf{b} = \begin{bmatrix} 0 \\ 1 \\ 3 \end{bmatrix}
$$ 
Hitung luas segitiga yang dibentuk oleh vektor 𝑎 dan 𝑏.

```python
import numpy as np

# Definisi vektor a dan b
a = np.array([2, 1, 0])
b = np.array([0, 1, 3])

# Hitung cross product a x b
cross_ab = np.cross(a, b)

# Hitung panjang (norma) dari hasil cross product
magnitude = np.linalg.norm(cross_ab)

# Luas segitiga adalah setengah dari panjang vektor hasil cross product
area_triangle = 0.5 * magnitude

# Tampilkan hasil
print("Vektor a × b =", cross_ab)
print("Magnitude dari a × b =", magnitude)
print("Luas segitiga =", area_triangle)
```

### 1.1.4 LINES AND PLANES

#### 1. VECTOR EQUATION OF A LINE

Garis dalam ruang tiga dimensi dapat direpresentasikan secara vektor sebagai:
$$ \mathbf{p}(t) = \mathbf{p}_1 + t\mathbf{d} $$

dengan

$ \mathbf{p}_1 $ : Titik yang dilalui garis.

$ \mathbf{D} $ : Vektor arah garis.

$t \in \mathbb{R}$ : Parameter bebas.

Jika garis melalui dua titik $ \mathbf{p}_1 $  dan $ \mathbf{p}_2 $, maka:
$$ \mathbf{p}(t) = \mathbf{p}_1 + t(\mathbf{p}_2 - \mathbf{p}_1) $$



Rumus parametris:
$$ x = x_1 + d_x t,\quad y = y_1 + d_y t,\quad z = z_1 + d_z t
$$


Rumus simetris (jika semua komponen $ \mathbf{d} \ne 0 $):
$$ \frac{x - x_1}{d_x} = \frac{y - y_1}{d_y} = \frac{z - z_1}{d_z}
$$
